### **Part 1: The Bias of the MLE Covariance and Bessel's Correction**

To prove that the raw Maximum Likelihood Estimator ($\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}$) is biased, we must evaluate the expectation of the sum of squared deviations from the sample mean.

**1. Expanding the Sum of Squares**
Let us start with the core summation inside the estimators. We can express the deviation of any single observation from the sample mean as the difference between its deviation from the true mean $\boldsymbol{\mu}$ and the sample mean's deviation from $\boldsymbol{\mu}$:
$$\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n = (\mathbf{X}_i - \boldsymbol{\mu}) - (\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})$$

Substitute this into the sum of the outer products:
$$\sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T = \sum_{i=1}^n [(\mathbf{X}_i - \boldsymbol{\mu}) - (\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})][(\mathbf{X}_i - \boldsymbol{\mu}) - (\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})]^T$$

Expanding this product yields four terms. Because $\sum_{i=1}^n (\mathbf{X}_i - \boldsymbol{\mu}) = n(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})$, the middle cross-terms simplify, condensing the entire expression to:
$$\sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T = \sum_{i=1}^n (\mathbf{X}_i - \boldsymbol{\mu})(\mathbf{X}_i - \boldsymbol{\mu})^T - n(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T$$

**2. Taking the Expectation**
Now, apply the linear expectation operator to both sides:
$$\mathbb{E}\left[ \sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T \right] = \sum_{i=1}^n \mathbb{E}[(\mathbf{X}_i - \boldsymbol{\mu})(\mathbf{X}_i - \boldsymbol{\mu})^T] - n\mathbb{E}[(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T]$$

We rely on two fundamental variance properties for i.i.d. variables:
* The variance of the population: $\mathbb{E}[(\mathbf{X}_i - \boldsymbol{\mu})(\mathbf{X}_i - \boldsymbol{\mu})^T] = \boldsymbol{\Sigma}$
* The variance of the sample mean: $\mathbb{E}[(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T] = \frac{1}{n}\boldsymbol{\Sigma}$

Substituting these standard properties back into our expectation:
$$\mathbb{E}\left[ \sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T \right] = \sum_{i=1}^n \boldsymbol{\Sigma} - n\left(\frac{1}{n}\boldsymbol{\Sigma}\right) = n\boldsymbol{\Sigma} - \boldsymbol{\Sigma} = (n-1)\boldsymbol{\Sigma}$$

**3. Proving MLE Bias**
The raw Maximum Likelihood Estimator divides this sum by $n$:
$$\widehat{\boldsymbol{\Sigma}}_{\text{MLE}} = \frac{1}{n} \sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T$$

Applying our expectation result:
$$\mathbb{E}[\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}] = \frac{1}{n} \mathbb{E}\left[ \sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T \right] = \frac{n-1}{n}\boldsymbol{\Sigma}$$
Because $\frac{n-1}{n} \neq 1$ for finite sample sizes, $\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}$ inherently underestimates the true covariance matrix and is therefore strictly biased.

**4. Demonstrating Bessel's Correction**
To correct this bias, we apply a multiplicative factor—Bessel's correction—of $\frac{n}{n-1}$ to the MLE estimator, yielding the sample covariance matrix $\mathbf{S}$:
$$\mathbf{S} = \frac{n}{n-1} \widehat{\boldsymbol{\Sigma}}_{\text{MLE}} = \frac{1}{n-1} \sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T$$

Taking the expectation of this corrected estimator:
$$\mathbb{E}[\mathbf{S}] = \frac{1}{n-1} \mathbb{E}\left[ \sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T \right] = \frac{1}{n-1} [(n-1)\boldsymbol{\Sigma}] = \boldsymbol{\Sigma}$$
This demonstrates that $\mathbf{S}$ successfully acts as a strictly unbiased estimator of the true population covariance matrix $\boldsymbol{\Sigma}$.

### **Part 2: Hypothesis Testing in Continuous Structural Health Monitoring**

Under the null hypothesis ($H_0$), the structural asset operates in a completely healthy state (the status quo).

**Defining the Error Types**
* **Type I Error ($\alpha$):** A **False Positive**. In structural monitoring, this occurs when the diagnostic system incorrectly rejects the null hypothesis and triggers an alarm indicating structural failure, even though the asset is perfectly healthy. This typically leads to unnecessary plant shutdowns and costly inspections.
* **Type II Error ($\beta$):** A **False Negative**. This occurs when the diagnostic system fails to reject the null hypothesis despite actual underlying damage. The monitoring system reports the structure is "healthy," allowing a compromised asset to continue operating and risking catastrophic mechanical failure.

**The Consequence of an Ultra-Conservative Threshold**
If an engineer sets an ultra-conservative significance threshold like $\alpha = 0.0001$ to rigorously avoid false alarms, this decision directly impacts the **Statistical Power** ($1 - \beta$) of the test.

Mathematically, $\alpha$ and $\beta$ are inversely related. By setting $\alpha$ exceptionally low, the critical boundary for the test statistic is pushed out to the extreme far tails of the null distribution. To reject $H_0$, a recorded signal must be overwhelmingly anomalous. Because it requires such an extreme signal to trigger an alarm, the probability of missing moderate or early-stage structural damage ($\beta$) increases drastically. Consequently, the statistical power—the probability of correctly identifying actual damage—plummets. The system becomes "deaf" to subtle faults.

**Geometrical Impact on the Confidence Ellipsoid**
In multivariate statistics, the boundary of normal operation is frequently defined by a Mahalanobis distance threshold, creating an $m$-dimensional confidence ellipsoid representing the "status quo."

When $\alpha = 0.0001$, the diagnostic layer essentially declares that 99.99% of the expected variance under healthy conditions must be treated as "normal operation." Geometrically, this forces the volume of the "healthy operation" confidence ellipsoid to expand massively. Because the ellipsoid is so expansive, early-stage structural deterioration will likely generate data vectors that still fall *inside* the boundary. The damage must be severe enough to push the data completely out of this bloated geometric region before the algorithm will register an anomaly.

### **1. Slutsky’s Theorem (Multivariate Formulation)**

Slutsky’s Theorem extends the algebraic properties of limits to sequences of random variables, allowing us to combine sequences that converge in distribution with sequences that converge in probability. 

**Formal Statement:**
Let $\mathbf{X}_n$ be a sequence of random vectors in $\mathbb{R}^m$ such that it converges in distribution to a random vector $\mathbf{X}$:
$$\mathbf{X}_n \xrightarrow{d} \mathbf{X}$$

Let $\mathbf{A}_n$ be a sequence of random matrices and $\mathbf{B}_n$ be a sequence of random vectors that converge in probability to a constant matrix $\mathbf{A}$ and a constant vector $\mathbf{b}$, respectively:
$$\mathbf{A}_n \xrightarrow{p} \mathbf{A} \quad \text{and} \quad \mathbf{B}_n \xrightarrow{p} \mathbf{b}$$

Then, Slutsky's Theorem states that:
1. $\mathbf{X}_n + \mathbf{B}_n \xrightarrow{d} \mathbf{X} + \mathbf{b}$
2. $\mathbf{A}_n \mathbf{X}_n \xrightarrow{d} \mathbf{A} \mathbf{X}$
3. $\mathbf{A}_n^{-1} \mathbf{X}_n \xrightarrow{d} \mathbf{A}^{-1} \mathbf{X}$ (provided that $\mathbf{A}_n$ and $\mathbf{A}$ are invertible).

---

### **2. Justifying the Practical Parametric Distribution**

To operationalize the distribution of the sample mean without knowing the latent population covariance matrix $\boldsymbol{\Sigma}$, we must prove that substituting it with the empirical sample covariance matrix $\mathbf{S}$ maintains the asymptotic Gaussian properties.

**Step 1: Establishing Convergence of the Covariance Matrix**
By the Weak Law of Large Numbers (WLLN), the unbiased empirical sample covariance matrix $\mathbf{S}$ is a consistent estimator for the true population covariance matrix. As the sample size approaches infinity:
$$\mathbf{S} \xrightarrow{p} \boldsymbol{\Sigma}$$

**Step 2: Applying the Continuous Mapping Theorem**
Because the matrix inverse and square root operations are continuous for positive definite matrices, we apply the Continuous Mapping Theorem to our convergence in probability result:
$$\mathbf{S}^{-1/2} \xrightarrow{p} \boldsymbol{\Sigma}^{-1/2}$$

**Step 3: Integrating the Lindeberg-Lévy MCLT**
The Lindeberg-Lévy Multivariate Central Limit Theorem provides our baseline convergence in distribution:
$$\sqrt{n}(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu}) \xrightarrow{d} \mathscr{N}(\mathbf{0}, \boldsymbol{\Sigma})$$

**Step 4: Applying Slutsky's Theorem**
We now have a sequence of random vectors converging in distribution ($\mathbf{X}_n$) and a sequence of random matrices converging in probability ($\mathbf{A}_n$). Let:
* $\mathbf{X}_n = \sqrt{n}(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})$
* $\mathbf{A}_n = \mathbf{S}^{-1/2}$
* $\mathbf{A} = \boldsymbol{\Sigma}^{-1/2}$

According to the multiplicative rule of Slutsky's Theorem ($\mathbf{A}_n \mathbf{X}_n \xrightarrow{d} \mathbf{A} \mathbf{X}$):
$$\mathbf{S}^{-1/2} \sqrt{n}(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu}) \xrightarrow{d} \boldsymbol{\Sigma}^{-1/2} \mathscr{N}(\mathbf{0}, \boldsymbol{\Sigma})$$

**Step 5: Simplifying the Asymptotic Distribution**
A linear transformation $\mathbf{C}\mathbf{Z}$ of a multivariate normal vector $\mathbf{Z} \sim \mathscr{N}(\mathbf{0}, \boldsymbol{\Sigma})$ results in a new normal distribution $\mathscr{N}(\mathbf{0}, \mathbf{C}\boldsymbol{\Sigma}\mathbf{C}^T)$. Applying this property to the right side of our limit:
$$\boldsymbol{\Sigma}^{-1/2} \mathscr{N}(\mathbf{0}, \boldsymbol{\Sigma}) \equiv \mathscr{N}\left(\mathbf{0}, \boldsymbol{\Sigma}^{-1/2}\boldsymbol{\Sigma}(\boldsymbol{\Sigma}^{-1/2})^T\right) = \mathscr{N}(\mathbf{0}, \mathbf{I})$$

Therefore, the standardized, sample-estimated statistic converges to a standard multivariate normal distribution:
$$\mathbf{S}^{-1/2} \sqrt{n}(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu}) \xrightarrow{d} \mathscr{N}(\mathbf{0}, \mathbf{I})$$

**Step 6: Deriving the Practical Operational Distribution**
To make this result useful for engineering or statistical modeling (where we require the distribution of $\widehat{\boldsymbol{\mu}}_n$ itself), we rearrange the finite-sample approximation of this asymptotic limit. 

For sufficiently large $n$, we undo the standardization. Multiplying the standard normal by $\mathbf{S}^{1/2}$ yields a variance of $\mathbf{S}^{1/2}\mathbf{I}(\mathbf{S}^{1/2})^T = \mathbf{S}$:
$$\sqrt{n}(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu}) \overset{\text{approx}}{\sim} \mathscr{N}(\mathbf{0}, \mathbf{S})$$

Dividing the variable by $\sqrt{n}$ scales the covariance matrix by $\frac{1}{n}$, and shifting by the constant vector $\boldsymbol{\mu}$ translates the mean:
$$\widehat{\boldsymbol{\mu}}_n \sim \mathscr{N}\left({\boldsymbol{\mu}}, \frac{1}{n}\mathbf{S}\right)$$

This confirms that Slutsky’s Theorem mathematically permits substituting the unobservable $\boldsymbol{\Sigma}$ with the empirical $\mathbf{S}$ in large samples, yielding a tractable, data-driven probability distribution.

````python
import numpy as np
import pandas as pd
import scipy.stats as stats


np.random.seed(42)
n_samples = 5000
n_features = 3


base_data = np.random.multivariate_normal(
    mean=[0.5, -0.2, 1.1],
    cov=[[0.09, 0.02, 0.01],
         [0.02, 0.06, 0.03],
         [0.01, 0.03, 0.05]],
    size=n_samples
)


base_data[4000:, 0] += 0.015  # Sensor 1 drift
base_data[4000:, 2] -= 0.010  # Sensor 3 drift

df_strain = pd.DataFrame(base_data, columns=['Strain_Ch1', 'Strain_Ch2', 'Strain_Ch3'])
# ----------------------------------------------------------------

def verify_first_moment_homogeneity(df: pd.DataFrame, g_chunks: int = 5) -> dict:
    """
    Partitions the dataset into g chunks and evaluates first-moment homogeneity
    via Wilks' Lambda and Bartlett's Chi-Square asymptotic transformation.
    """
    
    n, m = df.shape
    g = g_chunks
    n_j = n // g
    
    
    grand_mean = df.mean(axis=0).values
    W = np.zeros((m, m))
    B = np.zeros((m, m))
    
    # 3. Calculate
    for k in range(g):
        # Extract the current temporal block
        chunk = df.iloc[k*n_j : (k+1)*n_j].values
        chunk_mean = chunk.mean(axis=0)
        
        # Between-Chunk Sum of Squares (Weighted by chunk size)
        mean_diff = (chunk_mean - grand_mean).reshape(m, 1)
        B += n_j * (mean_diff @ mean_diff.T)
        
        # Within-Chunk Sum of Squares
        centered_chunk = chunk - chunk_mean
        W += centered_chunk.T @ centered_chunk
        
   
    T = B + W
    wilks_lambda = np.linalg.det(W) / np.linalg.det(T)
    

    df_chi2 = m * (g - 1)
    bartlett_multiplier = -(n - 1 - (m + g) / 2.0)
    chi2_stat = bartlett_multiplier * np.log(wilks_lambda)
    
    # Calculate p-value 
    p_value = stats.chi2.sf(chi2_stat, df_chi2)
    
    return {
        "Wilks_Lambda": wilks_lambda,
        "Bartlett_Chi2": chi2_stat,
        "Degrees_of_Freedom": df_chi2,
        "p_value": p_value
    }


results = verify_first_moment_homogeneity(df_strain, g_chunks=5)

print("--- MANOVA Diagnostic Results ---")
for key, value in results.items():
    if key == "Degrees_of_Freedom":
        print(f"{key:20}: {value}")
    else:
        print(f"{key:20}: {value:.6f}")
````

### **Part 1: The Covariance of the Transformed Vector**

**1. Mathematical Proof**
We are given the centered vector $\widetilde{\mathbf{X}}_i = (\mathbf{X}_i - \boldsymbol{\mu})$ and the projection $\mathbf{Z}_i = \mathbf{P}^T \widetilde{\mathbf{X}}_i$. To find the theoretical covariance matrix of $\mathbf{Z}_i$, we calculate the expectation of its outer product:

$$\mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T] = \mathbb{E}[(\mathbf{P}^T \widetilde{\mathbf{X}}_i)(\mathbf{P}^T \widetilde{\mathbf{X}}_i)^T]$$

Apply the matrix transpose property $(\mathbf{A}\mathbf{B})^T = \mathbf{B}^T \mathbf{A}^T$:
$$\mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T] = \mathbb{E}[\mathbf{P}^T \widetilde{\mathbf{X}}_i \widetilde{\mathbf{X}}_i^T \mathbf{P}]$$

Because the eigenvector matrix $\mathbf{P}$ is a constant linear transformation (it does not contain random variables), we can factor it outside the linear expectation operator:
$$\mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T] = \mathbf{P}^T \mathbb{E}[\widetilde{\mathbf{X}}_i \widetilde{\mathbf{X}}_i^T] \mathbf{P}$$

By definition, the expectation of the centered outer product is the population covariance matrix, $\mathbb{E}[\widetilde{\mathbf{X}}_i \widetilde{\mathbf{X}}_i^T] = \boldsymbol{\Sigma}$:
$$\mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T] = \mathbf{P}^T \boldsymbol{\Sigma} \mathbf{P}$$

Substitute the spectral decomposition of the covariance matrix ($\boldsymbol{\Sigma} = \mathbf{P} \mathbf{\Lambda} \mathbf{P}^T$):
$$\mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T] = \mathbf{P}^T (\mathbf{P} \mathbf{\Lambda} \mathbf{P}^T) \mathbf{P}$$

Because $\mathbf{P}$ is an orthogonal matrix, its transpose is its inverse ($\mathbf{P}^T \mathbf{P} = \mathbf{I}$):
$$\mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T] = \mathbf{I} \mathbf{\Lambda} \mathbf{I} = \mathbf{\Lambda}$$

**2. Geometric and Statistical Meaning**
* **Statistical Meaning:** The off-diagonal elements of $\mathbf{\Lambda}$ are all zero. This proves that the cross-covariances between any two distinct principal components $Z_{i,j}$ and $Z_{i,k}$ (where $j \neq k$) are precisely zero. The transformed variables are now perfectly linearly uncorrelated.
* **Geometric Meaning:** The projection matrix $\mathbf{P}^T$ acts as a rigid rotation of the coordinate system. It rotates the original basis vectors so that the new axes align exactly with the directions of maximum variance (the principal axes) of the $m$-dimensional data cloud, stripping away the linear dependencies between the original sensor channels.

### **Part 2: Total Variance and Subspace Ratios**

**1. Proof of Total Variance Preservation**
The total variance in the raw coordinate space is the sum of the variances of individual variables, defined as the trace of the covariance matrix, $\text{tr}(\boldsymbol{\Sigma})$.

Using the spectral decomposition $\boldsymbol{\Sigma} = \mathbf{P} \mathbf{\Lambda} \mathbf{P}^T$, we write:
$$\text{tr}(\boldsymbol{\Sigma}) = \text{tr}(\mathbf{P} \mathbf{\Lambda} \mathbf{P}^T)$$

The trace operator enjoys the property of cyclic permutation invariance, meaning $\text{tr}(\mathbf{A}\mathbf{B}\mathbf{C}) = \text{tr}(\mathbf{C}\mathbf{A}\mathbf{B})$. Applying this to shift the trailing $\mathbf{P}^T$ to the front:
$$\text{tr}(\boldsymbol{\Sigma}) = \text{tr}(\mathbf{P}^T \mathbf{P} \mathbf{\Lambda})$$

Since $\mathbf{P}$ is orthogonal ($\mathbf{P}^T \mathbf{P} = \mathbf{I}$):
$$\text{tr}(\boldsymbol{\Sigma}) = \text{tr}(\mathbf{I} \mathbf{\Lambda}) = \text{tr}(\mathbf{\Lambda})$$

Because $\mathbf{\Lambda}$ is a diagonal matrix containing the eigenvalues $\lambda_j$ along its trace, the sum simplifies exactly to:
$$\text{tr}(\boldsymbol{\Sigma}) = \sum_{j=1}^m \lambda_j$$
This mathematically guarantees that projecting data into the principal component space perfectly preserves the total "energy" (variance) of the system.

**2. Variance Ratio Formulations**
When retaining $k$ principal components out of $m$, the variance explained by the partitioned subspaces is rigorously defined as:

* **Cumulative Explained Variance Ratio ($\Phi(k)$):**
    $$\Phi(k) = \frac{\sum_{j=1}^k \lambda_j}{\sum_{j=1}^m \lambda_j}$$
* **Residual Unexplained Variance Ratio ($\Psi(k)$):**
    $$\Psi(k) = \frac{\sum_{j=k+1}^m \lambda_j}{\sum_{j=1}^m \lambda_j} = 1 - \Phi(k)$$

    $$\mathbf{e}_i = \mathbf{x}_i - \widehat{\mathbf{x}}_i = \widehat{\mathbf{P}}_{m-k} \mathbf{z}_{i,m-k}$$

**2. Pythagorean Decomposition Proof**
Starting with the partitioned centered vector $\widetilde{\mathbf{x}}_i = \widehat{\mathbf{P}}_k \mathbf{z}_{i,k} + \widehat{\mathbf{P}}_{m-k} \mathbf{z}_{i,m-k}$, the squared Euclidean norm is the inner product of the vector with itself:
$$\|\widetilde{\mathbf{x}}_i\|^2 = \widetilde{\mathbf{x}}_i^T \widetilde{\mathbf{x}}_i = \left(\widehat{\mathbf{P}}_k \mathbf{z}_{i,k} + \widehat{\mathbf{P}}_{m-k} \mathbf{z}_{i,m-k}\right)^T \left(\widehat{\mathbf{P}}_k \mathbf{z}_{i,k} + \widehat{\mathbf{P}}_{m-k} \mathbf{z}_{i,m-k}\right)$$

Distributing the transpose and expanding the binomial:
$$\|\widetilde{\mathbf{x}}_i\|^2 = \left(\mathbf{z}_{i,k}^T \widehat{\mathbf{P}}_k^T + \mathbf{z}_{i,m-k}^T \widehat{\mathbf{P}}_{m-k}^T\right) \left(\widehat{\mathbf{P}}_k \mathbf{z}_{i,k} + \widehat{\mathbf{P}}_{m-k} \mathbf{z}_{i,m-k}\right)$$
$$\|\widetilde{\mathbf{x}}_i\|^2 = \mathbf{z}_{i,k}^T \left(\widehat{\mathbf{P}}_k^T \widehat{\mathbf{P}}_k\right) \mathbf{z}_{i,k} + \mathbf{z}_{i,m-k}^T \left(\widehat{\mathbf{P}}_{m-k}^T \widehat{\mathbf{P}}_{m-k}\right) \mathbf{z}_{i,m-k} + 2\mathbf{z}_{i,k}^T \left(\widehat{\mathbf{P}}_k^T \widehat{\mathbf{P}}_{m-k}\right) \mathbf{z}_{i,m-k}$$

We exploit the orthogonality of the loading matrices:
1.  $\widehat{\mathbf{P}}_k^T \widehat{\mathbf{P}}_k = \mathbf{I}_k$
2.  $\widehat{\mathbf{P}}_{m-k}^T \widehat{\mathbf{P}}_{m-k} = \mathbf{I}_{m-k}$
3.  $\widehat{\mathbf{P}}_k^T \widehat{\mathbf{P}}_{m-k} = \mathbf{0}$ (The principal subspace and residual subspace are strictly orthogonal).

Substituting these identities annihilates the cross-term:
$$\|\widetilde{\mathbf{x}}_i\|^2 = \mathbf{z}_{i,k}^T \mathbf{z}_{i,k} + \mathbf{z}_{i,m-k}^T \mathbf{z}_{i,m-k}$$
$$\|\mathbf{x}_i - \widehat{\boldsymbol{\mu}}_n\|^2 = \|\mathbf{z}_{i,k}\|^2 + \|\mathbf{z}_{i,m-k}\|^2$$

**3. SHM Diagnostics: Hotelling's $T^2$ vs. $Q$ Statistic**
* **Extreme Environmental Wind Loads $\rightarrow$ Hotelling's $T^2$ Statistic**
    * *Justification:* The $T^2$ statistic measures variation *inside* the established PCA model plane. When wind loads increase, the structure bends heavily but the physics of its movement remain intact. The correlation structure between sensors does not break; the amplitude simply scales up along expected axes. $T^2$ will spike because it tracks extreme movements along known, healthy variance pathways.
* **Internal Structural Fracture $\rightarrow$ $Q$ Statistic (SPE)**
    * *Justification:* The $Q$ statistic measures the residual variance *orthogonal* to the PCA model plane. A fatigue crack changes the physical stiffness and mechanical load pathways of the structure. This causes sensors to suddenly drift out of phase and break their historical correlation mapping. Because this is a totally new mode of deformation, it cannot be reconstructed by the healthy $\widehat{\mathbf{P}}_k$ basis vectors. The data vector "lifts" off the normal plane, causing the residual projection error ($Q$) to spike.

## **Part B: Numerical Verification**



***Scenario:** You are building a real-time diagnostics layer for a structural component monitored by $m = 4$ sensors. You have a baseline archive of $n = 3,000$ nominal snapshots. To establish operational boundaries, you must write an optimization pipeline that analyzes how Hotelling’s $T^2$ and the residual $Q$ statistic respond to different truncation boundaries $k \in \{1, 2, 3\}$.*

Complete the Python class method below inside Google Colab. The function must:

1. Standardize and center the incoming realizations.

3. Loop through every possible subspace cutoff value $k \in \{1, 2, 3\}$, calculating the sample vector array for Hotelling’s $T^2$ ($T^2_i = \sum_{j=1}^k \frac{z_{i,j}^2}{\widehat{\lambda}_j}$) and the $Q$ statistic ($Q_i = \sum_{j=k+1}^m z_{i,j}^2$).

````python

# This command tells your Jupyter/VS Code environment to install the math libraries if they are missing.
%pip install numpy pandas scipy 

import numpy as np
import pandas as pd


# Simulating a baseline archive of n = 3,000 snapshots across m = 4 sensors
np.random.seed(42)
n_samples = 3000
m_sensors = 4

# Generating correlated structural sensor data to simulate healthy operational behavior
cov_matrix = [
    [1.0, 0.8, 0.5, 0.2],
    [0.8, 1.0, 0.7, 0.4],
    [0.5, 0.7, 1.0, 0.6],
    [0.2, 0.4, 0.6, 1.0]
]
sensor_data = np.random.multivariate_normal(mean=[0, 0, 0, 0], cov=cov_matrix, size=n_samples)
df_sensors = pd.DataFrame(sensor_data, columns=['Sensor_1', 'Sensor_2', 'Sensor_3', 'Sensor_4'])


class SHMDiagnostics:
    def __init__(self, data: np.ndarray):
        self.data = data
        self.n, self.m = data.shape

    def evaluate_truncation_boundaries(self):
        """
        Executes the PCA pipeline and calculates E[T^2] and E[Q] for k in {1, 2, 3}.
        """
        # Step 1: Standardize and center the incoming realizations (ddof=1 for sample std)
        mu_hat = np.mean(self.data, axis=0)
        std_hat = np.std(self.data, axis=0, ddof=1)
        X_centered = (self.data - mu_hat) / std_hat

        # Step 2: Decompose the unbiased sample covariance matrix S
        S = np.cov(X_centered, rowvar=False)
        eigenvalues, eigenvectors = np.linalg.eigh(S)

        # Sort components in descending order
        sorted_indices = np.argsort(eigenvalues)[::-1]
        lambda_hat = eigenvalues[sorted_indices]
        P_hat = eigenvectors[:, sorted_indices]

        # Project into principal component space: Z = X_centered * P_hat
        Z = X_centered @ P_hat

        results = {}

        # Step 3: Loop through every possible subspace cutoff value k in {1, 2, 3}
        for k in [1, 2, 3]:
            
            # Isolate the retained subspace (k components)
            Z_k = Z[:, :k]
            lambda_k = lambda_hat[:k]
            
            # Isolate the discarded residual subspace (m - k components)
            Z_residual = Z[:, k:]

            # Calculate Hotelling's T^2 array
            T2_array = np.sum((Z_k**2) / lambda_k, axis=1)

            # Calculate Residual Q statistic array (SPE)
            Q_array = np.sum(Z_residual**2, axis=1)

            # Step 4: Compute expected values (empirical means)
            E_T2 = np.mean(T2_array)
            E_Q = np.mean(Q_array)

            # Calculate theoretical expectations to verify Part A
            Theo_T2 = float(k)
            Theo_Q = np.sum(lambda_hat[k:])

            results[f'k={k}'] = {
                'Empirical_E[T2]': E_T2,
                'Theoretical_E[T2]': Theo_T2,
                'Empirical_E[Q]': E_Q,
                'Theoretical_E[Q]': Theo_Q
            }

        return results, lambda_hat


# Initialize and run the pipeline
pipeline = SHMDiagnostics(df_sensors.values)
metrics, evals = pipeline.evaluate_truncation_boundaries()

# Print the Final Output
print("\n" + "="*50)
print(f"Computed Eigenvalues: {evals.round(4)}")
print("="*50)

for k_val, values in metrics.items():
    print(f"\n--- Truncation Boundary: {k_val} ---")
    print(f"Empirical E[T^2]   : {values['Empirical_E[T2]']:.4f}  |  Theoretical E[T^2] : {values['Theoretical_E[T2]']:.4f}")
    print(f"Empirical E[Q]     : {values['Empirical_E[Q]']:.4f}  |  Theoretical E[Q]   : {values['Theoretical_E[Q]']:.4f}")

print("\n" + "="*50)
print("DIAGNOSTIC CONCLUSION:")
print("1. T^2 matches theoretical expectation (E[T^2] = k).")
print("2. Q matches the sum of the discarded eigenvalues.")
print("="*50)
````

### **Part 1: The Fundamental Equation of Factor Analysis**

**1. Proof of the Fundamental Equation**
We are given the linear generative structure $\mathbf{Z}_i = \boldsymbol{\Lambda} \mathbf{F}_i + \boldsymbol{\epsilon}_i$. To find the population covariance matrix $\mathbf{R}$, we evaluate the expectation of the outer product of $\mathbf{Z}_i$ with itself:

$$\mathbf{R} = \mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T]$$

Substitute the generative model into the expectation:
$$\mathbf{R} = \mathbb{E}[(\boldsymbol{\Lambda} \mathbf{F}_i + \boldsymbol{\epsilon}_i)(\boldsymbol{\Lambda} \mathbf{F}_i + \boldsymbol{\epsilon}_i)^T]$$

Expand the matrix product, utilizing the property $(\mathbf{A} + \mathbf{B})^T = \mathbf{A}^T + \mathbf{B}^T$ and $(\mathbf{A}\mathbf{B})^T = \mathbf{B}^T\mathbf{A}^T$:
$$\mathbf{R} = \mathbb{E}[\boldsymbol{\Lambda} \mathbf{F}_i \mathbf{F}_i^T \boldsymbol{\Lambda}^T + \boldsymbol{\Lambda} \mathbf{F}_i \boldsymbol{\epsilon}_i^T + \boldsymbol{\epsilon}_i \mathbf{F}_i^T \boldsymbol{\Lambda}^T + \boldsymbol{\epsilon}_i \boldsymbol{\epsilon}_i^T]$$

Because the expectation operator is strictly linear, we can distribute it and pull the constant loading matrices ($\boldsymbol{\Lambda}$) outside:
$$\mathbf{R} = \boldsymbol{\Lambda} \mathbb{E}[\mathbf{F}_i \mathbf{F}_i^T] \boldsymbol{\Lambda}^T + \boldsymbol{\Lambda} \mathbb{E}[\mathbf{F}_i \boldsymbol{\epsilon}_i^T] + \mathbb{E}[\boldsymbol{\epsilon}_i \mathbf{F}_i^T] \boldsymbol{\Lambda}^T + \mathbb{E}[\boldsymbol{\epsilon}_i \boldsymbol{\epsilon}_i^T]$$

Now, we substitute our defined assumptions:
1.  **Orthogonal Common Factors:** $\mathbb{E}[\mathbf{F}_i \mathbf{F}_i^T] = \mathbf{I}_{k \times k}$
2.  **Factor-Noise Independence:** $\mathbb{E}[\boldsymbol{\epsilon}_i \mathbf{F}_i^T] = \mathbf{0}$ (and its transpose $\mathbb{E}[\mathbf{F}_i \boldsymbol{\epsilon}_i^T] = \mathbf{0}$)
3.  **Diagonalized Noise:** $\mathbb{E}[\boldsymbol{\epsilon}_i \boldsymbol{\epsilon}_i^T] = \boldsymbol{\Psi}$

Applying these constraints annihilates the middle cross-terms:
$$\mathbf{R} = \boldsymbol{\Lambda} \mathbf{I}_{k \times k} \boldsymbol{\Lambda}^T + \mathbf{0} + \mathbf{0} + \boldsymbol{\Psi}$$
$$\mathbf{R} = \boldsymbol{\Lambda}\boldsymbol{\Lambda}^T + \boldsymbol{\Psi}$$

**2. Communality and Uniqueness in Diagnostics**
Looking at the $j$-th diagonal element of this equation ($R_{jj} = 1$ for a normalized matrix), we get $1 = h_j^2 + \varphi_j^2$.
* **Communality ($h_j^2 = \sum_{p=1}^k \lambda_{jp}^2$):** This represents the proportion of variance in the $j$-th sensor that is shared with other sensors and explained by the underlying latent factors. In asset diagnostics, a high communality means the sensor is responding to global system behaviors (e.g., overarching mechanical load, system-wide thermal expansion).
* **Uniqueness ($\varphi_j^2$):** This represents the proportion of variance completely unique to the $j$-th sensor. In diagnostics, this captures localized noise, micro-environmental effects directly at the sensor's mounting location, or independent sensor degradation/faults that do not impact the rest of the structure.

---

### **Part 2: Orthogonal Rotations in Factor Analysis**

**1. PCA Loadings vs. Varimax-Rotated FA Loadings**
* **Raw PCA Loadings:** PCA strictly forces the first principal component to maximize the total explained variance of the entire system. Consequently, the first PC loading vector is usually a "general" component where almost all sensors show moderate-to-high loadings. The subsequent components are purely mathematical orthogonal corrections, making them extremely difficult to map to isolated physical phenomena.
* **Varimax FA Loadings:** Varimax abandons the "maximize total variance per component" rule in favor of "simple structure." It rotates the axes to maximize the variance of the squared loadings within each column. This forces sensor loadings toward $0$ or extreme values ($1$ or $-1$). Conceptually, it isolates the variables: Factor 1 might represent *only* the thermal sensors, while Factor 2 represents *only* the vibration sensors, radically improving diagnostic interpretability.

**2. Preservation of Communality and Global Covariance**
An orthogonal rotation matrix $\mathbf{T}$ satisfies the property $\mathbf{T}\mathbf{T}^T = \mathbf{T}^T\mathbf{T} = \mathbf{I}$.
Let our new rotated loading matrix be $\tilde{\boldsymbol{\Lambda}} = \boldsymbol{\Lambda}\mathbf{T}$. If we substitute this back into our fundamental equation approximation:
$$\tilde{\boldsymbol{\Lambda}}\tilde{\boldsymbol{\Lambda}}^T + \boldsymbol{\Psi} = (\boldsymbol{\Lambda}\mathbf{T})(\boldsymbol{\Lambda}\mathbf{T})^T + \boldsymbol{\Psi}$$
$$\tilde{\boldsymbol{\Lambda}}\tilde{\boldsymbol{\Lambda}}^T + \boldsymbol{\Psi} = \boldsymbol{\Lambda}(\mathbf{T}\mathbf{T}^T)\boldsymbol{\Lambda}^T + \boldsymbol{\Psi}$$
$$\tilde{\boldsymbol{\Lambda}}\tilde{\boldsymbol{\Lambda}}^T + \boldsymbol{\Psi} = \boldsymbol{\Lambda}\mathbf{I}\boldsymbol{\Lambda}^T + \boldsymbol{\Psi} = \boldsymbol{\Lambda}\boldsymbol{\Lambda}^T + \boldsymbol{\Psi}$$
Because $\boldsymbol{\Lambda}\boldsymbol{\Lambda}^T$ evaluates to the exact same global matrix regardless of the orthogonal rotation $\mathbf{T}$, the global covariance approximation remains unchanged. Furthermore, because the entire matrix $\boldsymbol{\Lambda}\boldsymbol{\Lambda}^T$ is preserved, its specific diagonal elements (the communalities $h_j^2$) are perfectly mathematically preserved. The rotation merely redistributes the explained variance *among* the factors, not the total amount of variance explained.

### **Part 3: Thomson’s Regression Method for Factor Scores**

**1. Expansion of the Joint Covariance Matrix**
We define the partitioned random vector $\mathbf{Y}_i = \begin{bmatrix} \mathbf{Z}_i \\ \mathbf{F}_i \end{bmatrix}$. Because both $\mathbf{Z}_i$ and $\mathbf{F}_i$ are mean-centered, their expected values are zero. The joint population covariance matrix is the expectation of the outer product:

$$\boldsymbol{\Sigma}_{YY} = \mathbb{E}[\mathbf{Y}_i \mathbf{Y}_i^T] = \mathbb{E}\left[ \begin{bmatrix} \mathbf{Z}_i \\ \mathbf{F}_i \end{bmatrix} \begin{bmatrix} \mathbf{Z}_i^T & \mathbf{F}_i^T \end{bmatrix} \right]$$
$$\boldsymbol{\Sigma}_{YY} = \begin{bmatrix} \mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T] & \mathbb{E}[\mathbf{Z}_i \mathbf{F}_i^T] \\ \mathbb{E}[\mathbf{F}_i \mathbf{Z}_i^T] & \mathbb{E}[\mathbf{F}_i \mathbf{F}_i^T] \end{bmatrix}$$

We evaluate the four blocks:
1.  **Top-Left:** $\mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T] = \mathbf{R}$ (by definition).
2.  **Bottom-Right:** $\mathbb{E}[\mathbf{F}_i \mathbf{F}_i^T] = \mathbf{I}_{k \times k}$ (by definition of orthogonal factors).
3.  **Top-Right:** To find the cross-covariance, we substitute the generative model:
    $$\mathbb{E}[\mathbf{Z}_i \mathbf{F}_i^T] = \mathbb{E}[(\boldsymbol{\Lambda}\mathbf{F}_i + \boldsymbol{\epsilon}_i)\mathbf{F}_i^T] = \boldsymbol{\Lambda}\mathbb{E}[\mathbf{F}_i\mathbf{F}_i^T] + \mathbb{E}[\boldsymbol{\epsilon}_i\mathbf{F}_i^T]$$
    $$\mathbb{E}[\mathbf{Z}_i \mathbf{F}_i^T] = \boldsymbol{\Lambda}\mathbf{I} + \mathbf{0} = \boldsymbol{\Lambda}$$
4.  **Bottom-Left:** By symmetry, this is the transpose of the top-right block, $\boldsymbol{\Lambda}^T$.

Assembling the blocks yields the exact expanded matrix:
$$\boldsymbol{\Sigma}_{YY} = \begin{bmatrix} \mathbf{R} & \boldsymbol{\Lambda} \\ \boldsymbol{\Lambda}^T & \mathbf{I}_{k \times k} \end{bmatrix}$$

**2. Deriving the Final Score Matrix Equation**
We want to estimate the latent factors $\mathbf{F}_i$ given our observed sensor data $\mathbf{Z}_i = \mathbf{z}_i$. 
Using the provided conditional expectation theorem for multivariate normals:
$$\mathbb{E}[\mathbf{X}_2 | \mathbf{x}_1] = \boldsymbol{\mu}_2 + \boldsymbol{\Sigma}_{21} \boldsymbol{\Sigma}_{11}^{-1} (\mathbf{x}_1 - \boldsymbol{\mu}_1)$$

We map our partitioned $\boldsymbol{\Sigma}_{YY}$ variables to this theorem:
* $\mathbf{X}_2 = \mathbf{F}_i$ and $\mathbf{x}_1 = \mathbf{z}_i$
* Because the data is normalized, the baseline means are zero: $\boldsymbol{\mu}_2 = \mathbf{0}$ and $\boldsymbol{\mu}_1 = \mathbf{0}$
* From our partitioned covariance matrix, the data auto-covariance is $\boldsymbol{\Sigma}_{11} = \mathbf{R}$
* The cross-covariance from data to factors is $\boldsymbol{\Sigma}_{21} = \boldsymbol{\Lambda}^T$

Substituting these into the theorem:
$$\mathbf{f}_i = \mathbf{0} + \boldsymbol{\Lambda}^T \mathbf{R}^{-1} (\mathbf{z}_i - \mathbf{0})$$
$$\mathbf{f}_i = \boldsymbol{\Lambda}^T \mathbf{R}^{-1} \mathbf{z}_i$$
This is Thomson’s Regression Estimator, demonstrating that the unobservable latent factors can be linearly mapped from the observed sensor readings using the inverted global correlation matrix and the transposed factor loadings.

````python

%pip install numpy pandas scipy scikit-learn plotly nbformat

import numpy as np
import pandas as pd
from sklearn.decomposition import FactorAnalysis
import plotly.graph_objects as go
from plotly.subplots import make_subplots


# Re-creating the baseline 4-channel sensor array from Part A
np.random.seed(42)
n_samples = 3000
m_sensors = 4

cov_matrix = [
    [1.0, 0.8, 0.5, 0.2],
    [0.8, 1.0, 0.7, 0.4],
    [0.5, 0.7, 1.0, 0.6],
    [0.2, 0.4, 0.6, 1.0]
]
sensor_data = np.random.multivariate_normal(mean=[0, 0, 0, 0], cov=cov_matrix, size=n_samples)
df_sensors = pd.DataFrame(sensor_data, columns=['Sensor_1', 'Sensor_2', 'Sensor_3', 'Sensor_4'])


class FactorPartitionEngine:
    def __init__(self, data_frame: pd.DataFrame, target_factors: int = 2):
        self.raw_data = data_frame.select_dtypes(include=[np.number]).values
        self.features = data_frame.select_dtypes(include=[np.number]).columns.tolist()
        self.n, self.m = self.raw_data.shape
        self.k = target_factors
        
        # Extracted FA properties
        self.Z = None
        self.Lambda = None
        self.communality = None
        self.uniqueness = None
        self.factor_scores = None
        self.factor_variances = None

    def execute_mathematical_decomposition(self):
        """
        Applies mathematical guardrails, standardizes the metrics, fits a Varimax-rotated 
        Factor Analysis model, and projects scores using Thomson's regression.
        """
        # --- DIMENSIONALITY CONSTRAINT LAYER ---
        if self.n <= self.m:
            raise ValueError(f"CRITICAL: Snapshot count (n={self.n}) must exceed channel count (m={self.m}).")
        if not (self.k < self.m):
            raise ValueError(f"CRITICAL: Latent dimension (k={self.k}) must be strictly less than feature dimension (m={self.m}).")

        # --- Z-SCORE MAPPING LAYER ---
        mu = np.mean(self.raw_data, axis=0)
        stdevs = np.std(self.raw_data, axis=0, ddof=1)
        
        # Explicit guardrail for zero-variance channels
        stdevs = np.where(stdevs == 0, 1e-15, stdevs)
        self.Z = (self.raw_data - mu) / stdevs

        # --- INFORMATION DECOMPOSITION LAYER ---
        # Fit FA with Varimax rotation 
        fa = FactorAnalysis(n_components=self.k, rotation='varimax', random_state=42)
        fa.fit(self.Z)
        
        # Extract loading matrix transposed to match math notation (m x k)
        self.Lambda = fa.components_.T 
        
        # Compute Communality (h^2) and Uniqueness (phi^2)
        self.communality = np.sum(self.Lambda**2, axis=1)
        self.uniqueness = 1.0 - self.communality  # Assuming normalized variance = 1

        # LATENT SCORE PROJECTION (THOMSON'S METHOD) 
        # F = Z * R^(-1) * Lambda
        R = np.cov(self.Z, rowvar=False, ddof=1)
        R_inv = np.linalg.inv(R)
        self.factor_scores = self.Z @ R_inv @ self.Lambda
        
        # Compute empirical variance of extracted factors
        self.factor_variances = np.var(self.factor_scores, axis=0, ddof=1)

        # --- TELEMETRY CONSOLE PRINTOUT ---
        avg_communality = np.mean(self.communality) * 100
        avg_uniqueness = np.mean(self.uniqueness) * 100
        print(f"\n--- SYSTEM DIAGNOSTIC TELEMETRY ---")
        print(f"Target Latent Factors (k)    : {self.k}")
        print(f"Average System Communality % : {avg_communality:.2f}%")
        print(f"Average System Uniqueness %  : {avg_uniqueness:.2f}%\n")

    def render_diagnostic_dashboard(self):
        """
        Renders the strict 2x2 Plotly diagnostic grid layout.
        """
        factor_labels = [f"Factor_{i+1}" for i in range(self.k)]
        
        fig = make_subplots(
            rows=2, cols=2, 
            horizontal_spacing=0.24, 
            vertical_spacing=0.28,
            subplot_titles=(
                "Structural Loadings Matrix |Λ|", 
                "Variance Partitioning Profile", 
                "Sensor Uniqueness Noise Floor", 
                "Latent Factor Empirical Variance"
            )
        )

        # 1. Structural Loadings Heatmap (1, 1)
        fig.add_trace(go.Heatmap(
            z=np.abs(self.Lambda), x=factor_labels, y=self.features,
            colorscale='YlOrRd',
            colorbar=dict(x=-0.15, len=0.38, y=0.78, yanchor="middle", xanchor="right")
        ), row=1, col=1)

        # 2. Variance Partitioning Horizontal Stacked Bar (1, 2)
        fig.add_trace(go.Bar(
            y=self.features, x=self.communality * 100, 
            orientation='h', name='Shared Communality (h²)', marker_color='#1f77b4'
        ), row=1, col=2)
        fig.add_trace(go.Bar(
            y=self.features, x=self.uniqueness * 100, 
            orientation='h', name='Isolated Uniqueness (φ²)', marker_color='#ff7f0e'
        ), row=1, col=2)

        # 3. Sensor Uniqueness Line Profile (2, 1)
        fig.add_trace(go.Scatter(
            x=self.features, y=self.uniqueness, 
            mode='lines+markers', name='Noise Threshold',
            line=dict(color='#d62728', width=2, dash='dashdot'),
            marker=dict(symbol='x', color='#d62728', size=8)
        ), row=2, col=1)

        # 4. Latent Factor Variance Distribution (2, 2)
        fig.add_trace(go.Bar(
            x=factor_labels, y=self.factor_variances, 
            name='Empirical Variance',
            marker=dict(color='#2ca02c', line=dict(color='black', width=0.5))
        ), row=2, col=2)

        # --- ENFORCE GRID LAYOUT & GEOMETRIC RULES ---
        fig.update_layout(
            width=1250, height=750, 
            template='plotly_white', barmode='stack',
            legend=dict(orientation="h", x=0.5, y=1.02, xanchor="center", yanchor="bottom"),
            margin=dict(t=150, b=60, l=140, r=80),
            title=dict(text="<b>Parametric Subspace FA Diagnostics (Varimax Rotation)</b>", x=0.5, y=0.98, font=dict(size=20))
        )

        # Force strict axis limitations and rotations
        fig.update_xaxes(range=[0, 100], title_text="Variance %", row=1, col=2)
        fig.update_xaxes(tickangle=25, row=2, col=1)
        fig.update_yaxes(title_text="Uniqueness Fraction (φ²)", row=2, col=1)
        fig.update_yaxes(title_text="Variance", row=2, col=2)
        
        # Ensure correct y-axis reading for horizontal bars and heatmaps
        fig.update_yaxes(autorange="reversed", row=1, col=1)
        fig.update_yaxes(autorange="reversed", row=1, col=2)

        # THIS IS THE FIX: Bypassing VS Code's internal viewer to render in your web browser
        fig.show(renderer="browser")


# We truncate to k=2 common latent dimensions for this structure
diagnostics = FactorPartitionEngine(df_sensors, target_factors=2)
diagnostics.execute_mathematical_decomposition()
diagnostics.render_diagnostic_dashboard()
````

class FeatureAnalysisDashboard:
    def __init__(self, data_frame: pd.DataFrame):
        self.raw_data = data_frame.select_dtypes(include=[np.number]).values
        self.features = data_frame.select_dtypes(include=[np.number]).columns.tolist()
        self.n, self.m = self.raw_data.shape
        
        # Standardize Data (Z-Score Mapping)
        mu = np.mean(self.raw_data, axis=0)
        stdevs = np.std(self.raw_data, axis=0, ddof=1)
        stdevs = np.where(stdevs == 0, 1e-15, stdevs) # Zero-variance guardrail
        self.Z = (self.raw_data - mu) / stdevs

    def render_pca_suite(self):
        """
        UI Layout 1: PCA Optimization Suite (2x3 Grid Layout)
        """
        # --- PCA MATHEMATICS ---
        S = np.cov(self.Z, rowvar=False, ddof=1)
        eigenvalues, eigenvectors = np.linalg.eigh(S)
        
        # Sort descending
        idx = np.argsort(eigenvalues)[::-1]
        lambdas = eigenvalues[idx]
        P = eigenvectors[:, idx]
        
        # Variance allocations
        var_pct = (lambdas / np.sum(lambdas)) * 100
        cum_var = np.cumsum(var_pct)
        unexplained_var = 100.0 - cum_var
        
        # T^2 and Q Tracking Arrays
        Z_pca = self.Z @ P
        T2_means = []
        Q_means = []
        k_values = list(range(1, self.m + 1))
        
        for k in k_values:
            Z_k = Z_pca[:, :k]
            T2_array = np.sum((Z_k**2) / lambdas[:k], axis=1)
            T2_means.append(np.mean(T2_array))
            
            if k < self.m:
                Z_res = Z_pca[:, k:]
                Q_array = np.sum(Z_res**2, axis=1)
                Q_means.append(np.mean(Q_array))
            else:
                Q_means.append(0.0) # 0 variance leftover when k=m

        # --- PLOTLY UI 1: 2x3 GRID ---
        pc_labels = [f"PC_{i+1}" for i in range(self.m)]
        fig = make_subplots(
            rows=2, cols=3, 
            horizontal_spacing=0.1, vertical_spacing=0.2,
            subplot_titles=(
                "Feature Loadings Matrix |P|", "Absolute Eigenvalues (λ)", "Explained Variance %",
                "Residual Space (Unexplained %)", "Mean Hotelling's T² vs k", "Mean Q Statistic vs k"
            )
        )

        # 1,1: Feature Loadings Heatmap
        fig.add_trace(go.Heatmap(z=np.abs(P), x=pc_labels, y=self.features, colorscale='Blues', colorbar=dict(x=-0.1)), row=1, col=1)
        
        # 1,2: Absolute Eigenvalues Bar
        fig.add_trace(go.Bar(x=pc_labels, y=lambdas, marker_color='indigo', name='Eigenvalue'), row=1, col=2)
        
        # 1,3: Explained Variance (Bar + Line Overlay)
        fig.add_trace(go.Bar(x=pc_labels, y=var_pct, marker_color='teal', name='Marginal Var %'), row=1, col=3)
        fig.add_trace(go.Scatter(x=pc_labels, y=cum_var, mode='lines+markers', line=dict(dash='dash', color='orange'), name='Cumulative Var %'), row=1, col=3)
        
        # 2,1: Residual Space Bar
        fig.add_trace(go.Bar(x=pc_labels, y=unexplained_var, marker_color='firebrick', name='Unexplained %'), row=2, col=1)
        
        # 2,2: Mean T^2 Scatter Line
        fig.add_trace(go.Scatter(x=k_values, y=T2_means, mode='lines+markers', line=dict(color='purple'), name='T² Mean'), row=2, col=2)
        
        # 2,3: Mean Q Scatter Line
        fig.add_trace(go.Scatter(x=k_values, y=Q_means, mode='lines+markers', line=dict(color='red'), name='Q Mean'), row=2, col=3)

        fig.update_layout(
            title="<b>UI Layout 1: PCA Optimization Suite</b>", 
            height=750, width=1500, template='plotly_white', showlegend=False,
            margin=dict(t=120, b=50, l=120, r=50)
        )
        fig.update_yaxes(autorange="reversed", row=1, col=1)
        fig.show(renderer="browser")

### **Part C: Diagnostic Analysis & Theoretical Responses**

---

#### **Answer 1: The Total Variance Illusion**

1. **Physical Meaning of ~100% Uniqueness ($\varphi^2$):** A Uniqueness value exceeding 98% means that almost all of the data fluctuation recorded by `Sensor_4` is completely isolated to that specific channel. It shares virtually no covariance with the rest of the structural network. Physically, this indicates that `Sensor_4` is not measuring the global dynamics of the asset; it is measuring its own localized noise, an electrical fault, or severe environmental interference specific to its mounting location.

2. **How Localized Noise "Tricks" PCA:**
PCA is a purely geometric algorithm designed to maximize *total global variance*. It is completely blind to whether that variance is "useful" shared structural movement or "useless" isolated noise. Because `Sensor_4` has a massive variance magnitude ($\sigma^2 \approx 2.0$ compared to the structural modes), PCA will aggressively orient one of its primary eigenvectors (often PC1 or PC2) directly toward the `Sensor_4` axis just to capture that large mathematical spread. 

3. **The Risk to Automated Anomaly Detection:**
If an engineer relies solely on PCA, the most "important" features (top principal components) in the diagnostic model will actually be modeling the broken sensor's noise rather than the true structural health of the plant. If a genuine, subtle structural fracture occurs in the sub-systems monitored by Sensors 1, 2, or 3, the anomaly detection system might miss it entirely because the global PCA threshold is artificially inflated by Sensor 4’s massive noise floor.

---

#### **Answer 2: Decoupling Structural Loading via Rotation**

1. **Varimax Rotation vs. Strict Mathematical Hierarchy:**
Raw PCA forces its axes to be strictly orthogonal and mandates that PC1 captures the absolute maximum variance possible, followed by PC2, etc. This mathematical greediness creates "dense" loading vectors where every sensor mixes into every principal component. Varimax rotation abandons the "maximize total variance per component" rule. Instead, it rotates the orthogonal axes in space to maximize the variance of the *squared loadings within each column*. This forces the weights toward $1$ (strong connection) or $0$ (no connection), explicitly mapping specific clusters of sensors to specific, isolated latent factors (creating "simple structure").

2. **Troubleshooting Advantage for Plant Operators:**
With a raw PCA matrix, if $T^2$ spikes, the operator cannot easily identify the root cause because all sensors contribute to the component. With a Varimax-rotated FA matrix, the factors map directly to physical reality. Because `Sensor_1` and `Sensor_2` load heavily on Factor 1, and `Sensor_3` loads exclusively on Factor 2, an operator knows exactly what is breaking. If Factor 1 scores suddenly drift, the maintenance crew is dispatched specifically to the mechanical sub-system housing Sensors 1 and 2, completely ignoring Sensor 3. 

---

#### **Answer 3: Determining Subspace Truncation ($k$) using $T^2$ and $Q$ Profiles**

1. **Behavior of the Mean $Q$ Statistic Curve:**
* **$k=1 \rightarrow k=2$:** The Mean $Q$ Statistic drops sharply. At $k=1$, the model has completely discarded the secondary latent mode ($f_2$), leaving massive amounts of real, structured physical variance in the residual space. Expanding the boundary to $k=2$ absorbs $f_2$ into the model, rapidly deflating the residual error.
* **$k=2 \rightarrow k=3$:** The curve flattens out almost entirely. Moving to $k=3$ does not capture any new structural physics; it only absorbs random, unstructured measurement noise from the sensors.

2. **The "Elbow" as the True Physical Dimensionality:**
The distinct "elbow" at $k=2$ visually separates the *signal* from the *noise floor*. A steep slope indicates the subspace is still capturing genuine, correlated physical energy. A flat line indicates the subspace is starved of real physics and is now just eating into random static. Therefore, $k=2$ is the true rank of the underlying physical system.

3. **The Penalty for Over-estimation ($k=3$):**
If you set the truncation cutoff to $k=3$, you commit **overfitting**. You are forcing raw, random measurement noise (and likely the massive isolated fault data from `Sensor_4`) into your "clean" principal subspace. This pollutes your Hotelling's $T^2$ statistic with static, drastically increasing the risk of false alarms (Type I errors) during normal operation.

---

#### **Answer 4: Operational Trade-offs in System Health Monitoring**

1. **The PCA Strategy:** PCA is computationally lightweight and geometrically rigid. It is excellent for tracking the global boundaries of a highly correlated, perfectly healthy system where all sensors are reliable. The $Q$ statistic is incredibly sensitive to any break in historical correlation. However, PCA relies entirely on the assumption that total variance equals structural variance. 

2. **The FA Strategy:** Factor Analysis explicitly partitions the data into a shared correlation matrix ($\boldsymbol{\Lambda}\boldsymbol{\Lambda}^T$) and a dedicated diagonal noise matrix ($\boldsymbol{\Psi}$). It mathematically acknowledges that sensors have unique, localized behavior that has nothing to do with the global asset.

**Robustness Verdict:**
**The Factor Analysis strategy is vastly more robust** against a single sensor experiencing an electrical short or loss of calibration. 

If `Sensor_4` shorts out and its variance explodes, PCA will warp its entire global plane to account for the new variance, destroying the whole monitoring model. Factor Analysis, however, will recognize that `Sensor_4`'s new variance is not shared with Sensors 1, 2, or 3. The FA engine will dynamically push that explosive variance entirely into `Sensor_4`'s **Uniqueness Noise Floor ($\varphi^2$)**, keeping its shared communality ($h^2$) near $0$. The shared structural loading matrix ($\boldsymbol{\Lambda}$) remains stable and uncorrupted, allowing the system to keep monitoring the true health of the asset while simultaneously flagging `Sensor_4` as a localized hardware failure.
