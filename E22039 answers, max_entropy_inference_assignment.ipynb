# Q. Statistical Mechanics of the Quantum Oscillator

Consider a single quantum harmonic oscillator of angular frequency $\omega$ in thermal equilibrium with a heat bath at temperature $T$. The quantum mechanical energy eigenvalues are given by:

$$E_j = \hbar \omega \left(j + \frac{1}{2}\right), \quad \text{for } j = 0, 1, 2, \dots$$

where $\hbar$ is the reduced Planck constant and $k_B$ is the Boltzmann constant.

**Answer the Following:**

1. **The Partition Function:**
Show that the canonical partition function $Z(T) = \sum_{j=0}^\infty e^{-E_j / (k_B T)}$ forms a convergent geometric series that simplifies to:

$$Z(T) = \frac{e^{-\frac{\hbar \omega}{2k_BT}}}{1 - e^{-\frac{\hbar \omega}{k_BT}}}$$


2. **Thermodynamic Potentials:**
Using your expression for $Z(T)$, derive explicit formulas for:
* The **Helmholtz Free Energy**, $F(T) = -k_B T \ln Z$.
* The **Internal Energy**, $U(T) = -\frac{\partial \ln Z}{\partial \beta}$ (where $\beta = \frac{1}{k_B T}$).


3. **Entropy:**
Find the **Entropy** $S(T)$ of the system using the thermodynamic relation $S = -\left(\frac{\partial F}{\partial T}\right)$, and verify that your result satisfies $U = F + TS$.
4. **Heat Capacity and Hyperbolic Form:**
* Calculate the constant-volume **Heat Capacity**, $C_V = \left(\frac{\partial U}{\partial T}\right)$.
* Use the hyperbolic identity $\sinh(x) = \frac{e^x - e^{-x}}{2}$ to prove that your expression for $C_V$ can be compactly rewritten as:

$$C_V = k_B \left(\frac{\hbar \omega}{2k_BT}\right)^2 \frac{1}{\sinh^2\left(\frac{\hbar \omega}{2k_BT}\right)}$$




5. **Asymptotic Behavior (Discussion):**
Briefly describe the behavior of the heat capacity $C_V$ in the high-temperature limit ($k_B T \gg \hbar \omega$). Does it match the classical equipartition theorem prediction?

### **1. The Partition Function**

The canonical partition function is given by the sum over all states:
$$Z(T) = \sum_{j=0}^\infty e^{-\beta E_j}$$
where $\beta = \frac{1}{k_B T}$ and the energy levels are $E_j = \hbar \omega \left(j + \frac{1}{2}\right)$.

Substitute the energy eigenvalues into the sum:
$$Z(T) = \sum_{j=0}^\infty e^{-\beta \hbar \omega \left(j + \frac{1}{2}\right)}$$

Factor out the constant zero-point energy term $e^{-\frac{\beta \hbar \omega}{2}}$:
$$Z(T) = e^{-\frac{\beta \hbar \omega}{2}} \sum_{j=0}^\infty e^{-\beta \hbar \omega j} = e^{-\frac{\beta \hbar \omega}{2}} \sum_{j=0}^\infty \left(e^{-\beta \hbar \omega}\right)^j$$

The sum is an infinite geometric series of the form $\sum_{j=0}^\infty x^j = \frac{1}{1-x}$ where $x = e^{-\beta \hbar \omega}$. Since $\beta \hbar \omega > 0$, we know $x < 1$, so the series converges to:
$$Z(T) = \frac{e^{-\frac{\hbar \omega}{2k_BT}}}{1 - e^{-\frac{\hbar \omega}{k_BT}}}$$

---

### **2. Thermodynamic Potentials**

**Helmholtz Free Energy ($F$)**
Using the formula $F = -k_B T \ln Z$:
$$F = -k_B T \ln \left( \frac{e^{-\frac{\beta \hbar \omega}{2}}}{1 - e^{-\beta \hbar \omega}} \right)$$
$$F = -k_B T \left[ \ln\left(e^{-\frac{\beta \hbar \omega}{2}}\right) - \ln\left(1 - e^{-\beta \hbar \omega}\right) \right]$$
$$F = -k_B T \left[ -\frac{\hbar \omega}{2k_B T} - \ln\left(1 - e^{-\frac{\hbar \omega}{k_B T}}\right) \right]$$
$$F(T) = \frac{\hbar \omega}{2} + k_B T \ln\left(1 - e^{-\frac{\hbar \omega}{k_B T}}\right)$$

**Internal Energy ($U$)**
First, write $\ln Z$ in terms of $\beta$:
$$\ln Z = -\frac{\beta \hbar \omega}{2} - \ln\left(1 - e^{-\beta \hbar \omega}\right)$$

Now take the derivative with respect to $\beta$:
$$U = -\frac{\partial \ln Z}{\partial \beta} = -\left[ -\frac{\hbar \omega}{2} - \frac{1}{1 - e^{-\beta \hbar \omega}} \cdot \left(\hbar \omega e^{-\beta \hbar \omega}\right) \right]$$
$$U = \frac{\hbar \omega}{2} + \frac{\hbar \omega e^{-\beta \hbar \omega}}{1 - e^{-\beta \hbar \omega}}$$

Multiply the numerator and denominator of the second term by $e^{\beta \hbar \omega}$ to simplify:
$$U(T) = \frac{\hbar \omega}{2} + \frac{\hbar \omega}{e^{\frac{\hbar \omega}{k_B T}} - 1}$$

---

### **3. Entropy**

Using the thermodynamic relation $S = -\left(\frac{\partial F}{\partial T}\right)$:
$$S = -\frac{\partial}{\partial T} \left[ \frac{\hbar \omega}{2} + k_B T \ln\left(1 - e^{-\frac{\hbar \omega}{k_B T}}\right) \right]$$

Apply the product rule to the second term:
$$S = - \left[ k_B \ln\left(1 - e^{-\frac{\hbar \omega}{k_B T}}\right) + k_B T \cdot \frac{1}{1 - e^{-\frac{\hbar \omega}{k_B T}}} \cdot \left(-e^{-\frac{\hbar \omega}{k_B T}}\right) \cdot \left(\frac{\hbar \omega}{k_B T^2}\right) \right]$$
$$S(T) = -k_B \ln\left(1 - e^{-\frac{\hbar \omega}{k_B T}}\right) + \frac{\hbar \omega}{T} \frac{1}{e^{\frac{\hbar \omega}{k_B T}} - 1}$$

**Verification ($U = F + TS$)**
Let's substitute our expressions into $F + TS$:
$$F + TS = \left[ \frac{\hbar \omega}{2} + k_B T \ln\left(1 - e^{-\frac{\hbar \omega}{k_B T}}\right) \right] + T \left[ -k_B \ln\left(1 - e^{-\frac{\hbar \omega}{k_B T}}\right) + \frac{\hbar \omega}{T} \frac{1}{e^{\frac{\hbar \omega}{k_B T}} - 1} \right]$$

The $k_B T \ln(\dots)$ terms cancel out perfectly:
$$F + TS = \frac{\hbar \omega}{2} + \frac{\hbar \omega}{e^{\frac{\hbar \omega}{k_B T}} - 1} = U(T)$$
This confirms the derivations are exact.

---

### **4. Heat Capacity and Hyperbolic Form**

**Calculating $C_V$**
$$C_V = \left(\frac{\partial U}{\partial T}\right) = \frac{\partial}{\partial T} \left[ \frac{\hbar \omega}{2} + \frac{\hbar \omega}{e^{\frac{\hbar \omega}{k_B T}} - 1} \right]$$

The derivative using the chain rule is:
$$C_V = \hbar \omega \cdot \left( -1 \right) \left(e^{\frac{\hbar \omega}{k_B T}} - 1\right)^{-2} \cdot \frac{\partial}{\partial T}\left(e^{\frac{\hbar \omega}{k_B T}}\right)$$
$$C_V = -\frac{\hbar \omega}{\left(e^{\frac{\hbar \omega}{k_B T}} - 1\right)^2} \cdot e^{\frac{\hbar \omega}{k_B T}} \cdot \left(-\frac{\hbar \omega}{k_B T^2}\right)$$
$$C_V = \frac{(\hbar \omega)^2 e^{\frac{\hbar \omega}{k_B T}}}{k_B T^2 \left(e^{\frac{\hbar \omega}{k_B T}} - 1\right)^2}$$

**Converting to Hyperbolic Form**
Focus on the fractional part of $C_V$ and divide the numerator and denominator by $e^{\frac{\hbar \omega}{k_B T}}$:
$$\frac{e^{\beta \hbar \omega}}{\left(e^{\beta \hbar \omega} - 1\right)^2} = \frac{1}{e^{-\beta \hbar \omega} \left(e^{\beta \hbar \omega} - 1\right)^2} = \frac{1}{\left(e^{\frac{\beta \hbar \omega}{2}} - e^{-\frac{\beta \hbar \omega}{2}}\right)^2}$$

Recall the identity $\sinh(x) = \frac{e^x - e^{-x}}{2}$, which means $4\sinh^2(x) = (e^x - e^{-x})^2$.
Setting $x = \frac{\beta \hbar \omega}{2}$, we get:
$$\left(e^{\frac{\beta \hbar \omega}{2}} - e^{-\frac{\beta \hbar \omega}{2}}\right)^2 = 4\sinh^2\left(\frac{\hbar \omega}{2k_B T}\right)$$

Substitute this back into the $C_V$ equation:
$$C_V = \frac{(\hbar \omega)^2}{k_B T^2} \cdot \frac{1}{4\sinh^2\left(\frac{\hbar \omega}{2k_B T}\right)}$$
$$C_V = k_B \left(\frac{\hbar \omega}{2k_BT}\right)^2 \frac{1}{\sinh^2\left(\frac{\hbar \omega}{2k_BT}\right)}$$

---

### **5. Asymptotic Behavior (Discussion)**

In the **high-temperature limit** ($k_B T \gg \hbar \omega$), the argument inside the hyperbolic sine function becomes very small. Let $x = \frac{\hbar \omega}{2k_B T}$, where $x \ll 1$.

For small values of $x$, the Taylor expansion of $\sinh(x)$ is approximately $x$. Therefore, $\sinh^2(x) \approx x^2$.

Substituting this approximation into our $C_V$ equation:
$$C_V \approx k_B (x)^2 \frac{1}{x^2} = k_B$$

**Does this match the classical prediction?**
Yes. The classical equipartition theorem states that every quadratic degree of freedom contributes $\frac{1}{2} k_B T$ to the internal energy. A 1D harmonic oscillator has two such degrees of freedom: kinetic energy ($\propto p^2$) and potential energy ($\propto x^2$). 
Therefore, the classical internal energy is $U_{\text{classical}} = 2 \times \frac{1}{2} k_B T = k_B T$. 
Taking the derivative with respect to temperature gives $C_V = k_B$. The quantum mechanical result correctly converges to the exact classical limit at high temperatures.

# Q. The Maximum Entropy Distribution under Mean and Variance Constraints

Consider the random variable $X \in (-\infty, \infty)$. Let:

1. **The Mean:** $\mathbb{E}[X] = \mu$
2. **The Variance:** $\mathbb{E}[(X-\mu)^2] = \sigma^2$

Show that the law of $X$ that maximizes entropy is given by:

$$X \sim \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left( -\frac{(x - \mu)^2}{2\sigma^2} \right)$$

# Maximum Entropy Distribution (Mean and Variance Constraints)

First, let's write down the definition of differential entropy for a continuous random variable $X$ with probability density function $p(x)$:

$$H(p) = -\int_{-\infty}^\infty p(x) \ln p(x) \, dx$$

We are trying to maximize this entropy subject to three constraints:
1. **Total probability must be 1:** $$\int_{-\infty}^\infty p(x) \, dx = 1$$
2. **The mean is $\mu$:** $$\int_{-\infty}^\infty (x - \mu) p(x) \, dx = 0$$
3. **The variance is $\sigma^2$:** $$\int_{-\infty}^\infty (x - \mu)^2 p(x) \, dx = \sigma^2$$

---

### Setting up the Lagrangian
To handle these constraints, I'll use the method of Lagrange multipliers. Let's define a functional $\mathcal{L}[p]$ with multipliers $\lambda_0, \lambda_1, \lambda_2$:

$$\mathcal{L}[p] = -\int_{-\infty}^\infty p(x) \ln p(x) \, dx + \lambda_0 \left( \int_{-\infty}^\infty p(x) \, dx - 1 \right) + \lambda_1 \left( \int_{-\infty}^\infty (x - \mu) p(x) \, dx \right) + \lambda_2 \left( \int_{-\infty}^\infty (x - \mu)^2 p(x) \, dx - \sigma^2 \right)$$

Next, I'll take the functional derivative of $\mathcal{L}$ with respect to $p(x)$ and set it to zero to find the stationary point:

$$\frac{\delta \mathcal{L}}{\delta p(x)} = -1 - \ln p(x) + \lambda_0 + \lambda_1 (x - \mu) + \lambda_2 (x - \mu)^2 = 0$$

Rearranging this to solve for $p(x)$:
$$\ln p(x) = (\lambda_0 - 1) + \lambda_1(x - \mu) + \lambda_2(x - \mu)^2$$
$$p(x) = \exp \left( (\lambda_0 - 1) + \lambda_1(x - \mu) + \lambda_2(x - \mu)^2 \right)$$

---

### Finding the Lagrange Multipliers
Now I need to use the constraints to figure out the constants.

First, look at the mean constraint. If I substitute $y = x - \mu$, the distribution becomes $p(y+\mu) \propto \exp(\lambda_1 y + \lambda_2 y^2)$. 
For the integral $\int y p(y) dy = 0$ to hold, the function has to be symmetric around $y=0$. The $\lambda_1 y$ term ruins this symmetry, so the only way this works is if $\lambda_1 = 0$. 
Also, for the probability to integrate to 1 (not blow up to infinity), $\lambda_2$ must be negative. 

So, the PDF simplifies to:
$$p(x) = A \exp \left( \lambda_2(x - \mu)^2 \right)$$
*(where $A = e^{\lambda_0 - 1}$ is just a constant to normalize the function).*

Let's write $\lambda_2 = -c$ (with $c > 0$) to make the integrals easier. The two remaining constraints are:
1. $\int_{-\infty}^\infty A e^{-c(x - \mu)^2} \, dx = 1$
2. $\int_{-\infty}^\infty A (x - \mu)^2 e^{-c(x - \mu)^2} \, dx = \sigma^2$

Using standard Gaussian integrals:
The first constraint gives:
$$A \sqrt{\frac{\pi}{c}} = 1 \implies A = \sqrt{\frac{c}{\pi}}$$

The second constraint gives:
$$A \left( \frac{1}{2c} \sqrt{\frac{\pi}{c}} \right) = \sigma^2$$

Plugging $A$ into the second equation:
$$\sqrt{\frac{c}{\pi}} \left( \frac{1}{2c} \sqrt{\frac{\pi}{c}} \right) = \sigma^2$$
$$\frac{1}{2c} = \sigma^2 \implies c = \frac{1}{2\sigma^2}$$

Now, plug $c$ back into the equation for $A$:
$$A = \sqrt{\frac{1/(2\sigma^2)}{\pi}} = \frac{1}{\sqrt{2\pi\sigma^2}}$$

Substitute $A$ and $c$ back into $p(x)$ and we get the normal distribution:
$$p(x) = \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left( -\frac{(x - \mu)^2}{2\sigma^2} \right)$$

---

### Proving it's a Maximum (KL Divergence)
The calculus of variations just gives a stationary point. To prove this is actually the maximum possible entropy, I can use Kullback-Leibler (KL) divergence. 

Let $p(x)$ be the Gaussian we just found, and $q(x)$ be any other valid PDF with the same mean $\mu$ and variance $\sigma^2$. The KL divergence is always non-negative:
$$D_{KL}(q \parallel p) = \int_{-\infty}^\infty q(x) \ln \frac{q(x)}{p(x)} \, dx \geq 0$$

Let's expand this:
$$\int_{-\infty}^\infty q(x) \ln q(x) \, dx - \int_{-\infty}^\infty q(x) \ln p(x) \, dx \geq 0$$
$$-H(q) - \int_{-\infty}^\infty q(x) \left[ \ln\left(\frac{1}{\sqrt{2\pi\sigma^2}}\right) - \frac{(x - \mu)^2}{2\sigma^2} \right] dx \geq 0$$

Because $q(x)$ is constrained to have the exact same variance $\sigma^2$ and normalizes to 1, evaluating that right-side integral with $q(x)$ gives the exact same result as if we did it with $p(x)$:
$$-H(q) - \int_{-\infty}^\infty p(x) \ln p(x) \, dx \geq 0$$
$$-H(q) + H(p) \geq 0$$
$$H(p) \geq H(q)$$

Since $H(p) \geq H(q)$ for any other valid distribution $q(x)$, the Gaussian distribution $p(x)$ truly is the global maximum entropy distribution for a given mean and variance.

# Q. Multiclass Logistic Regression: Automatic Classification of Emails

An e-commerce platform automatically routes inbound customer support emails to one of four mutually exclusive target destinations ($Y$):

* $Y = 0$ : **Finance**
* $Y = 1$ : **Technical**

Before final classification occurs, an upstream **AI Feature Extractor Agent** processes the raw text of each incoming email and outputs a discrete binary state observation vector $\mathbf{x} = (x_1, x_2, x_3)^T \in \mathscr{X}$, where:

* $x_1 = \mathbb{I}(\text{Financial context detected})$ (e.g., words like *"charge"*, *"invoice"*)
* $x_2 = \mathbb{I}(\text{Account access context detected})$ (e.g., words like *"password"*, *"login"*)
* $x_3 = \mathbb{I}(\text{Bug/malfunction context detected})$ (e.g., words like *"error"*, *"frozen"*)

---

Because human language inherently overlaps, the AI Agent's outputs are distributed non-deterministically across three common operational scenarios. The marginal probability of observing each state is fixed by historical logs to:

1. **Scenario A (Pure Financial):** $\mathbf{x}_A = (1, 0, 0)^T \implies P(\mathbf{x}_A) = 0.40$
2. **Scenario B (Pure Acces):** $\mathbf{x}_B = (0, 1, 0)^T \implies P(\mathbf{x}_B) = 0.25$
3. **Scenario C (Pure Technical):** $\mathbf{x}_C = (0, 0, 1)^T \implies P(\mathbf{x}_C) = 0.05$
3. **Scenario D (Access Technical):** $\mathbf{x}_D = (0, 1, 1)^T \implies P(\mathbf{x}_D) = 0.15$
3. **Scenario E (Billing Access):** $\mathbf{x}_E = (1, 1, 0)^T \implies P(\mathbf{x}_E) = 0.05$

---

Historical manual audits provide the following observed joint dataset expectations:

* Constraint 1: $\mathbb{E}[x_1 \cdot \mathbb{I}(Y=1)] = \mathbf{0.05}$
(Very low. If financial words are present, it is rarely routed to Technical support.)

* Constraint 2: $\mathbb{E}[x_2 \cdot \mathbb{I}(Y=1)] = \mathbf{0.15}$
(Moderate. Account access issues split down the middle depending on context.)

* Constraint 3: $\mathbb{E}[x_3 \cdot \mathbb{I}(Y=1)] = \mathbf{0.16}$
(High. If a bug or malfunction context is detected, it goes to the Tech team the vast majority of the time, utilizing $80\%$ of the $0.20$ max capacity).




### **1. The Logistic Regression Model Setup**

Even though the prompt mentions four target destinations, only two are explicitly defined ($Y=0$ for Finance, $Y=1$ for Technical). Therefore, this reduces to a binary Logistic Regression (Maximum Entropy) problem. 

The probability that an email is routed to the Technical department given a feature vector $\mathbf{x}$ takes the standard sigmoid form:

$$P(Y=1 \mid \mathbf{x}) = \frac{1}{1 + \exp(-(w_1 x_1 + w_2 x_2 + w_3 x_3))} = \sigma(w_1 x_1 + w_2 x_2 + w_3 x_3)$$

Let $p_k = P(Y=1 \mid \mathbf{x}_k)$ be the model's predicted probability for each scenario. By plugging in the binary state vectors, we get:
* **Scenario A (Pure Financial):** $\mathbf{x}_A = (1, 0, 0)^T \implies p_A = \sigma(w_1)$
* **Scenario B (Pure Access):** $\mathbf{x}_B = (0, 1, 0)^T \implies p_B = \sigma(w_2)$
* **Scenario C (Pure Technical):** $\mathbf{x}_C = (0, 0, 1)^T \implies p_C = \sigma(w_3)$
* **Scenario D (Access Technical):** $\mathbf{x}_D = (0, 1, 1)^T \implies p_D = \sigma(w_2 + w_3)$
* **Scenario E (Billing Access):** $\mathbf{x}_E = (1, 1, 0)^T \implies p_E = \sigma(w_1 + w_2)$

---

### **2. Applying the Maximum Entropy Constraints**

To find the weights, the model's expected feature values must match the empirical dataset expectations (historical manual audits).

The expectation equation is:
$$\mathbb{E}[x_i \cdot \mathbb{I}(Y=1)] = \sum_{\mathbf{x}} P(\mathbf{x}) P(Y=1 \mid \mathbf{x}) x_i$$

Applying this to our three constraints creates our system of equations:

**Constraint 1: Financial Context ($x_1$)**
The feature $x_1 = 1$ only in Scenarios A and E.
$$P(\mathbf{x}_A) p_A + P(\mathbf{x}_E) p_E = 0.05$$
$$0.40 \sigma(w_1) + 0.05 \sigma(w_1 + w_2) = 0.05$$

**Constraint 2: Account Access Context ($x_2$)**
The feature $x_2 = 1$ in Scenarios B, D, and E.
$$P(\mathbf{x}_B) p_B + P(\mathbf{x}_D) p_D + P(\mathbf{x}_E) p_E = 0.15$$
$$0.25 \sigma(w_2) + 0.15 \sigma(w_2 + w_3) + 0.05 \sigma(w_1 + w_2) = 0.15$$

**Constraint 3: Bug/Malfunction Context ($x_3$)**
The feature $x_3 = 1$ in Scenarios C and D.
$$P(\mathbf{x}_C) p_C + P(\mathbf{x}_D) p_D = 0.16$$
$$0.05 \sigma(w_3) + 0.15 \sigma(w_2 + w_3) = 0.16$$

---

### **3. Solving the System**

We now have 3 non-linear equations with 3 unknown weights. Because the sigmoid function $\sigma(z)$ is non-linear, there is no closed-form algebraic solution. 

To solve this computationally, we use Python's optimization libraries (such as `scipy.optimize.fsolve` to find the roots, or `scipy.optimize.minimize` on a loss function) to calculate the exact numerical values for $w_1, w_2,$ and $w_3$.

# Q. Asset Pricing and Portfolio Optimization.



In standard finance, models often assume asset returns follow a perfect Normal (Gaussian) distribution. However, real-world markets experience extreme events, leading to "fat tails" and skewed returns. MaxEnt allows an analyst to infer the **least-biased probability distribution** of an asset's future price based *only* on what the market currently tells us, without forcing it into a rigid Gaussian shape.

**Background:**

You are a quantitative portfolio manager evaluating a highly volatile tech stock, **Asset X**, currently trading at $\$100$. You want to determine the risk-neutral probability distribution of its price, $S_T$, in exactly one month. This distribution will allow you to correctly price exotic derivatives and optimize your portfolio's risk exposure.

**Available Market Data (The Constraints):**

You do not know the true distribution, but you can observe the following macroeconomic aggregates and market-clearing prices from the options exchange:

1. **The Risk-Free Growth:** Based on the risk-free interest rate, the expected future price of the asset must match the forward price:

$$\mathbb{E}[S_T] = \$102$$


2. **The Market Volatility:** Based on standard historical variance, the expected squared price is constrained by:

$$\mathbb{E}[S_T^2] = 10,600$$


3. **Asymmetric Market Sentiment (The "Skew"):** Due to heavy trading on out-of-the-money put options, the market is pricing in a strong downside risk. The expected value of a specific logarithmic penalty (which captures downward crash risk) is constrained to:

$$\mathbb{E}[\ln(S_T)] = 4.60$$



**Objective:**

Formulate and solve the Maximum Entropy problem to find the continuous probability density function $p(S_T)$ for $S_T > 0$.



# Maximum Entropy Formulation: Asset Pricing with Skew

To find the least-biased probability distribution of the asset's future price $S_T$, we will use the calculus of variations to maximize the Shannon entropy subject to the market constraints.

For mathematical brevity, let $x = S_T$, where $x \in (0, \infty)$ since the stock price cannot be negative.

### 1. Formulating the Objective and Constraints

The differential entropy of the continuous distribution $p(x)$ is defined as:
$$H(p) = -\int_0^\infty p(x) \ln p(x) \, dx$$

We want to maximize $H(p)$ subject to four constraints derived from the market data:
1. **Normalization:** The probabilities must sum to 1.
   $$\int_0^\infty p(x) \, dx = 1$$
2. **Mean (Forward Price):** $$\int_0^\infty x p(x) \, dx = 102$$
3. **Second Moment (Volatility):** $$\int_0^\infty x^2 p(x) \, dx = 10,600$$
4. **Logarithmic Penalty (Downside Skew):** $$\int_0^\infty \ln(x) p(x) \, dx = 4.60$$

---

### 2. Constructing the Lagrangian

We incorporate these constraints into a functional $\mathcal{L}[p]$ using Lagrange multipliers $\lambda_0, \lambda_1, \lambda_2,$ and $\lambda_3$:

$$\mathcal{L}[p] = -\int_0^\infty p(x) \ln p(x) \, dx + \lambda_0 \left( \int_0^\infty p(x) \, dx - 1 \right) + \lambda_1 \left( \int_0^\infty x p(x) \, dx - 102 \right) + \lambda_2 \left( \int_0^\infty x^2 p(x) \, dx - 10,600 \right) + \lambda_3 \left( \int_0^\infty \ln(x) p(x) \, dx - 4.60 \right)$$

To find the distribution that maximizes entropy, we take the functional derivative of $\mathcal{L}$ with respect to $p(x)$ and set it to zero:

$$\frac{\delta \mathcal{L}}{\delta p(x)} = -1 - \ln p(x) + \lambda_0 + \lambda_1 x + \lambda_2 x^2 + \lambda_3 \ln(x) = 0$$

---

### 3. Deriving the Functional Form

We solve the derivative equation for $p(x)$:

$$\ln p(x) = (\lambda_0 - 1) + \lambda_1 x + \lambda_2 x^2 + \lambda_3 \ln(x)$$

Exponentiating both sides:

$$p(x) = \exp(\lambda_0 - 1) \cdot \exp(\lambda_1 x + \lambda_2 x^2) \cdot \exp(\lambda_3 \ln x)$$

Let the normalization constant (or partition function) be defined as $Z = \exp(1 - \lambda_0)$. Recognizing that $\exp(\lambda_3 \ln x) = x^{\lambda_3}$, the final risk-neutral probability density function is:

$$p(S_T) = \frac{1}{Z} S_T^{\lambda_3} \exp\left(\lambda_1 S_T + \lambda_2 S_T^2\right)$$

---

### 4. Interpretation and Solution Constraints

This result belongs to a **three-parameter exponential family** and elegantly combines elements of different standard distributions:
* The $S_T$ and $S_T^2$ terms are characteristic of a **Gaussian (Normal)** distribution, fitting the mean and variance.
* The $S_T^{\lambda_3}$ term is characteristic of a **Gamma** or **Log-Normal**-like behavior, which actively pulls the distribution to accommodate the asymmetric left-tail skew dictated by the options market.

**Boundary Conditions for the Multipliers:**
For this distribution to be a valid, integrable probability density over $(0, \infty)$:
* $\lambda_2 < 0$: To ensure the right tail decays to zero as $S_T \to \infty$.
* $\lambda_3 > -1$: To ensure the probability mass does not diverge to infinity as $S_T \to 0$.

**Numerical Solution:**
Because the integrals of this hybrid functional form do not yield closed-form algebraic equations for the constraints, there is no analytical solution for $\lambda_1, \lambda_2,$ and $\lambda_3$. In a real-world quantitative pipeline, an analyst would find these exact values computationally by minimizing the dual function (the negative log-likelihood) using a numerical optimization algorithm such as Newton-Raphson or Gradient Descent.
