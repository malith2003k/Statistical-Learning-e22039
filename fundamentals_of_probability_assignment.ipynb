# Q2) Bertrand's Paradox

A chord of a circle is chosen “at random.” What is the probability that its length exceeds the side length of the inscribed equilateral triangle?

Show that different mathematically natural models of “random chord” lead to different answers.

### Solution

This is the famous *Bertrand Paradox*, which shows that different reasonable definitions of a “random chord” give different probabilities.

Let the circle have radius $R$. The side length of the inscribed equilateral triangle is
\[
s = \sqrt{3}R.
\]

A chord has length greater than $s$ if and only if its perpendicular distance $d$ from the center satisfies
\[
d < \frac{R}{2}.
\]

---

### Model 1: Random endpoints on the circumference

Choose two points independently and uniformly on the circle.

Let $\theta$ be the angle between them (in radians). The chord length is
\[
L = 2R \sin\left(\frac{\theta}{2}\right).
\]

We want:
\[
2R \sin\left(\frac{\theta}{2}\right) > \sqrt{3}R
\]
\[
\sin\left(\frac{\theta}{2}\right) > \frac{\sqrt{3}}{2}.
\]

This gives:
\[
\frac{\theta}{2} > \frac{\pi}{3} \quad \Rightarrow \quad \theta > \frac{2\pi}{3}.
\]

Since $\theta$ is uniform in $[0,\pi]$,
\[
P = \frac{\pi - \frac{2\pi}{3}}{\pi} = \frac{1}{3}.
\]

---

### Model 2: Random radius, choose a point uniformly along it

Pick a radius and then choose a point uniformly along it. Draw the chord perpendicular to the radius at that point.

Here, the distance $d$ from the center is uniformly distributed on $[0,R]$.

We need:
\[
d < \frac{R}{2}.
\]

Thus,
\[
P = \frac{R/2}{R} = \frac{1}{2}.
\]

---

### Model 3: Random midpoint uniformly in the disk

Choose the midpoint of the chord uniformly over the area of the circle.

The chord length exceeds $s$ if:
\[
d < \frac{R}{2}.
\]

Now probability depends on area:
\[
P = \frac{\text{area of circle of radius } R/2}{\text{area of circle of radius } R}
= \frac{\pi (R/2)^2}{\pi R^2}
= \frac{1}{4}.
\]

---

### Conclusion

We obtain three different answers:
\[
\frac{1}{3}, \quad \frac{1}{2}, \quad \frac{1}{4}.
\]

These differ because “choosing a random chord” is ambiguous unless we specify the method.

Each model corresponds to a different probability distribution over the set of chords, leading to different results. This is the essence of Bertrand’s paradox.

#Q8) Probability of Being a Girl?

A family has two children. Assume each child is equally likely to be a boy or a girl, independently of the other.

What is the probability that both children are girls, given that **at least one** of them is a girl?

### Solution

Assume each child is equally likely to be a boy (B) or a girl (G), independently.

The sample space for two children is:
\[
\{BB, BG, GB, GG\}
\]
Each outcome has probability:
\[
\frac{1}{4}.
\]

---

### Step 1: Condition on “at least one is a girl”

The event “at least one girl” excludes the outcome $BB$.

So the conditional sample space becomes:
\[
\{BG, GB, GG\}
\]

Each of these outcomes is now equally likely with probability:
\[
\frac{1}{3}.
\]

---

### Step 2: Find the required probability

We want:
\[
P(\text{both girls} \mid \text{at least one girl})
\]

Only one outcome satisfies “both girls,” which is $GG$.

Thus,
\[
P = \frac{1}{3}.
\]

---

### Final Answer

\[
\boxed{\frac{1}{3}}
\]

# Probability Space and Random Variables

## 1. Probability Space

The sample space is:

$$
\Omega = \{G_1, G_2, D_1, D_2\}
$$

The sigma-algebra is:

$$
\mathfrak{F} = \mathcal{P}(\Omega)
$$

The probability measure is:

$$
P(G_1)=0.50,\quad P(G_2)=0.20,\quad P(D_1)=0.10,\quad P(D_2)=0.20
$$

---

## 2. Random Variables

### Random variable \(X\)

$$
X(G_1)=0,\quad X(G_2)=0,\quad X(D_1)=1,\quad X(D_2)=1
$$

### Random variable \(Y\)

$$
Y(G_1)=1,\quad Y(G_2)=2,\quad Y(D_1)=3,\quad Y(D_2)=4
$$

---

## 3. Sigma-algebra generated by \(X\)

Since \(X\) groups outcomes into good and defective:

$$
\sigma(X)=\{\emptyset,\Omega,\{G_1,G_2\},\{D_1,D_2\}\}
$$

---

## 4. Sigma-algebra generated by \(Y\)

Since \(Y\) uniquely identifies each outcome:

$$
\sigma(Y)=\mathcal{P}(\Omega)
$$

---

## 5. Relationship between \(\sigma(X)\) and \(\sigma(Y)\)

$$
\sigma(X)\subseteq \sigma(Y)
$$

Thus, \(Y\) contains more detailed information than \(X\).

---

## 6. Marginal Distribution of \(X\)

$$
P(X=0)=P(G_1)+P(G_2)=0.7
$$

$$
P(X=1)=P(D_1)+P(D_2)=0.3
$$

---

## 7. Marginal Distribution of \(Y\)

$$
P(Y=1)=0.50,\quad P(Y=2)=0.20,\quad P(Y=3)=0.10,\quad P(Y=4)=0.20
$$

---

## 8. Joint Probabilities

$$
P(X=0,Y=1)=0.50
$$

$$
P(X=0,Y=2)=0.20
$$

$$
P(X=1,Y=3)=0.10
$$

$$
P(X=1,Y=4)=0.20
$$

---

## 9. Conditional Probabilities

$$
P(X=1|Y=3)=1,\quad P(X=1|Y=4)=1
$$

$$
P(Y=3|X=1)=\frac{0.10}{0.30}=\frac{1}{3}
$$

$$
P(Y=4|X=1)=\frac{0.20}{0.30}=\frac{2}{3}
$$

---

## 10. Interpretation

- \(X\): only tells whether product is good or defective.
- \(Y\): gives full information (quality + machine).

Thus, \(Y\) is more informative than \(X\).

---

## 11. Entropy Calculations

### Entropy of \(X\)

$$
H(X) = -0.7\log_2(0.7) - 0.3\log_2(0.3)
$$

$$
H(X) \approx 0.881 \text{ bits}
$$

---

### Entropy of \(Y\)

$$
H(Y) = -\sum p(y)\log_2 p(y)
$$

$$
H(Y)=-(0.5\log_2 0.5 + 0.2\log_2 0.2 + 0.1\log_2 0.1 + 0.2\log_2 0.2)
$$

$$
H(Y) \approx 1.761 \text{ bits}
$$

---

### Joint Entropy

Since mapping is one-to-one:

$$
H(X,Y)=H(Y)\approx 1.761 \text{ bits}
$$

---

### Conditional Entropy

$$
H(X|Y)=0
$$

$$
H(Y|X)\approx 0.881 \text{ bits}
$$

---

## 12. Summary

$$
H(X)=0.881,\quad H(Y)=1.761
$$

$$
H(X|Y)=0,\quad H(Y|X)=0.881
$$
