# Q. Gaussian Mixture Clustering as Conditional Updating

Consider a dataset
$$
x_1,x_2,\dots,x_n\in\mathbb R^d.
$$
We wish to cluster these observations into $K$ groups. Instead of assigning each point deterministically to a cluster at the beginning, we introduce a latent random variable
$$
C_i\in{1,\dots,K},
$$
where $C_i=k$ means that the observation $x_i$ belongs to cluster $k$.
Let the prior probability of cluster membership be
$$
P(C_i=k)=\phi_k,
$$
where
$$
\phi_k\ge 0,
\qquad
\sum_{k=1}^K \phi_k=1.
$$

Conditional on $C_i=k$, assume that the observation $X_i$ is generated from a multivariate Gaussian distribution:
$$
X_i\mid C_i=k
\sim
\mathscr N(\mu_k,\Sigma_k),
$$
where
$$
\mu_k\in\mathbb R^d,
\qquad
\Sigma_k\in\mathbb R^{d\times d}
$$
are the mean vector and covariance matrix of cluster $k$.

The model parameters
$$
\phi_k,\mu_k,\Sigma_k,
\qquad k=1,\dots,K,
$$
are assumed to be fixed but unknown.

## 1. Deriving the Marginal Density

Using the law of total probability, the marginal density of an observation $X_i$ is found by summing the joint probability of the observation and the cluster assignment over all possible clusters:

$$p(x_i)=\sum_{k=1}^K P(X_i=x_i \mid C_i=k)P(C_i=k)$$

Substituting the given prior probability $P(C_i=k)=\phi_k$ and the conditional Gaussian distribution $P(X_i=x_i \mid C_i=k)=\mathscr{N}(x_i \mid \mu_k, \Sigma_k)$, we obtain:

$$p(x_i)=\sum_{k=1}^K \phi_k \mathscr{N}(x_i \mid \mu_k, \Sigma_k)$$

**Why it is called a Gaussian mixture density:**
This density is a linear combination (a "mixture") of $K$ individual Gaussian distributions. The parameters $\phi_k$ act as mixing weights that determine the relative proportion of each Gaussian component in the overall dataset, allowing the model to represent complex, multi-modal distributions.

---

## 2. Deriving the Posterior Cluster Probability

By applying Bayes' rule to the joint distribution of $X_i$ and $C_i$, the posterior probability of cluster $k$ given the observation $x_i$ is:

$$P(C_i=k \mid X_i=x_i)=\frac{P(X_i=x_i \mid C_i=k)P(C_i=k)}{p(x_i)}$$

Expanding the marginal denominator $p(x_i)$ using the law of total probability from Part 1 gives:

$$P(C_i=k \mid X_i=x_i)=\frac{P(X_i=x_i \mid C_i=k)P(C_i=k)}{\sum_{j=1}^K P(X_i=x_i \mid C_i=j)P(C_i=j)}$$

Substituting our known distributions yields the responsibility $\gamma_{ik}$:

$$\gamma_{ik}=P(C_i=k \mid X_i=x_i)=\frac{\phi_k \mathscr{N}(x_i \mid \mu_k, \Sigma_k)}{\sum_{j=1}^K \phi_j \mathscr{N}(x_i \mid \mu_j, \Sigma_j)}$$

**Interpretation as a posterior probability:**
Before observing a data point, our belief that it belongs to cluster $k$ is just the prior $\phi_k$. Once we observe its exact spatial location $x_i$, Bayes' rule updates our belief. The responsibility $\gamma_{ik}$ quantifies how likely it is that cluster $k$ generated this specific point, relative to all other clusters.

---

## 3. One-Hot Encoding of the Latent Cluster Variable

The variable $Z_{ik}$ is an indicator function (taking values $0$ or $1$). For any indicator variable, its expected value is simply the probability that it equals $1$.

Conditioned on the observation $x_i$:

$$\mathbb{E}[Z_{ik} \mid X_i=x_i]=1 \cdot P(Z_{ik}=1 \mid X_i=x_i) + 0 \cdot P(Z_{ik}=0 \mid X_i=x_i)$$

Since $Z_{ik}=1$ is strictly equivalent to $C_i=k$, we substitute to find:

$$\mathbb{E}[Z_{ik} \mid X_i=x_i]=P(C_i=k \mid X_i=x_i)=\gamma_{ik}$$

By applying this expectation element-wise to the entire vector $Z_i$, we get:

$$\mathbb{E}[Z_i \mid X_i=x_i]=\begin{bmatrix} \mathbb{E}[Z_{i1} \mid x_i] \\ \vdots \\ \mathbb{E}[Z_{iK} \mid x_i] \end{bmatrix} = \begin{bmatrix} \gamma_{i1} \\ \vdots \\ \gamma_{iK} \end{bmatrix}$$

**Conclusion:** The mathematical conditional expectation of the one-hot encoded label vector exactly equals the vector of probabilities across all $K$ clusters, which is the definition of a soft cluster assignment.

---

## 4. From Soft Assignment to Hard Clustering

**Soft clustering** assigns a probability distribution to each observation. A point isn't forced into one cluster; instead, it is assigned fractional memberships across all clusters based on $\gamma_{ik}$ (e.g., $80\%$ Cluster 1, $20\%$ Cluster 2).

**Hard clustering**, given by $\widehat{C}_i=\operatorname{arg\,max}_{1 \le k \le K} \gamma_{ik}$, forces a discrete, deterministic decision. It discards the uncertainty of the probability distribution and assigns the point $100\%$ to the single cluster that yields the highest posterior responsibility.

---

## 5. Conditional Expectation of the Observation Given the Cluster

By definition, if an observation comes from cluster $k$ ($C_i=k$), it is drawn from a Gaussian distribution with mean $\mu_k$ and covariance $\Sigma_k$. The expected value of a Gaussian distribution is its mean:

$$\mathbb{E}[X_i \mid C_i=k]=\mu_k$$

**Why $\mu_k$ is the center:**
The expected value represents the highest-density region and theoretical spatial center of mass for all data points generated by cluster $k$.

**Comparing the Expectations:**
*   $\mathbb{E}[Z_i \mid X_i=x_i]$ is a distribution over **categories (clusters)** given a fixed spatial location. It tells you *who generated the point*.
*   $\mathbb{E}[X_i \mid C_i=k]$ is a distribution over **physical space** given a fixed category. It tells you *where the cluster is located*.

---

## 6. The Complete-Data Likelihood

Starting from the complete-data likelihood equation (assuming independent observations):

$$p(x_1, \dots, x_n, z_1, \dots, z_n)=\prod_{i=1}^n \prod_{k=1}^K \left[ \phi_k \mathscr{N}(x_i \mid \mu_k, \Sigma_k) \right]^{z_{ik}}$$

Taking the natural logarithm of both sides converts the products into sums, and the exponent $z_{ik}$ drops down as a multiplier:

$$\ell_c=\log p(X, Z)=\sum_{i=1}^n \sum_{k=1}^K z_{ik} \log \left[ \phi_k \mathscr{N}(x_i \mid \mu_k, \Sigma_k) \right]$$

Applying logarithm rules ($\log(ab)=\log a + \log b$) yields:

$$\ell_c=\sum_{i=1}^n \sum_{k=1}^K z_{ik} \left[ \log \phi_k + \log \mathscr{N}(x_i \mid \mu_k, \Sigma_k) \right]$$

**Why this is easy to maximize:**
If the latent variables $z_{ik}$ were known, they would act as binary switches ($0$ or $1$). The total log-likelihood would elegantly decouple into $K$ independent maximization problems. We could update the parameters for cluster $k$ using only the specific data points where $z_{ik}=1$, reducing the problem to standard Gaussian maximum likelihood estimation.

---

## 7. The EM Interpretation

Because $z_{ik}$ is unknown, we cannot directly compute $\ell_c$. Instead, the E-step takes the expected value of $\ell_c$ given the observed data and current parameters. Because $\ell_c$ is a linear function of $z_{ik}$, we simply replace $z_{ik}$ with its conditional expectation, $\gamma_{ik}$:

$$Q=\sum_{i=1}^n \sum_{k=1}^K \gamma_{ik} \left[ \log \phi_k + \log \mathscr{N}(x_i \mid \mu_k, \Sigma_k) \right]$$

**Why this is a conditional update:**
The E-step essentially says, "We don't know the exact hard labels, so we will update our best guess of the cluster memberships ($\gamma_{ik}$) conditionally based on where the current cluster centers and shapes are."

---

## 8. Parameter Updates

By setting the partial derivatives of $Q$ with respect to $\phi_k$, $\mu_k$, and $\Sigma_k$ to zero (incorporating the constraint $\sum \phi_k=1$ via Lagrange multipliers), we arrive at the standard updates provided in the prompt.

**Interpretation of fractional weight:**
Because we don't use binary hard labels, $N_k$ represents the "effective" number of points in cluster $k$. During the M-step, instead of a data point fully defining one cluster's center and variance, it contributes a fraction of its coordinates weighted exactly by $\gamma_{ik}$. If $\gamma_{i1}=0.9$, point $x_i$ pulls cluster $1$'s mean $\mu_1$ with $90\%$ of its influence, and other clusters with only $10\%$.

---

## 9. Interpretation

Gaussian Mixture clustering can be viewed as a repeated process of conditional updating that bridges assumptions and evidence. The mixture weight $\phi_k$ serves as the prior probability of cluster $k$ before any data is seen. Once we observe data, the Gaussian density $\mathscr{N}(x_i \mid \mu_k, \Sigma_k)$ measures how spatially compatible $x_i$ is with cluster $k$'s current shape and location. Bayes' rule combines these to form the responsibility $\gamma_{ik}$, which is the posterior probability of cluster $k$ after observing $x_i$. This entire soft assignment vector is mathematically represented as $\mathbb{E}[Z_i \mid X_i=x_i]$. The M-step then updates the physical cluster parameters using these posterior membership probabilities as fractional weights, shifting the centers toward the points that heavily belong to them. Therefore, Gaussian mixture clustering is fundamentally probabilistic clustering based on iteratively computing the conditional expectations of latent cluster membership variables.

```python
# 10. Computational Simulation and Out-of-Sample Validation

import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
from sklearn.mixture import GaussianMixture
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split

class GMMFinancialSegmenter:
    def __init__(self, n_components=3, random_state=42):
        self.n_components = n_components
        self.random_state = random_state
        self.scaler = StandardScaler()
        self.gmm = GaussianMixture(
            n_components=self.n_components,
            random_state=self.random_state,
            max_iter=500
        )

    def prepare_data(self, df, feature_cols):
        """Scales features and splits into train/test sets."""
        self.feature_cols = feature_cols

        # Drop NaNs in target columns for clean modeling
        df_clean = df.dropna(subset=feature_cols)
        X = df_clean[feature_cols].values

        # Standardize features to prevent scale variance dominance
        X_scaled = self.scaler.fit_transform(X)

        self.X_train, self.X_test = train_test_split(
            X_scaled, test_size=0.2, random_state=self.random_state
        )
        print(f"Data prepared: {self.X_train.shape[0]} train rows, {self.X_test.shape[0]} test rows.")

    def fit_and_evaluate(self):
        """Executes the EM algorithm and validates on out-of-sample data."""
        self.gmm.fit(self.X_train)

        print("\n--- EM Algorithm Execution ---")
        print(f"Converged: {self.gmm.converged_}")
        print(f"Iterations required: {self.gmm.n_iter_}")

        # Out-of-Sample Performance
        train_ll = self.gmm.score(self.X_train)
        test_ll = self.gmm.score(self.X_test)

        print("\n--- Log-Likelihood Performance ---")
        print(f"Average Training Log-Likelihood: {train_ll:.4f}")
        print(f"Average Test Log-Likelihood:     {test_ll:.4f}")

    def _create_grid(self, X):
        """Creates a fine coordinate grid for contour mapping."""
        x_min, x_max = X[:, 0].min() - 0.5, X[:, 0].max() + 0.5
        y_min, y_max = X[:, 1].min() - 0.5, X[:, 1].max() + 0.5
        xx, yy = np.meshgrid(np.linspace(x_min, x_max, 200),
                             np.linspace(y_min, y_max, 200))
        return xx, yy, np.c_[xx.ravel(), yy.ravel()]

    def plot_empirical_density(self):
        """1. 2D Density Heatmap (Using Bulletproof Defaults)"""
        df_train = pd.DataFrame(self.X_train, columns=self.feature_cols)
        fig = px.density_heatmap(
            df_train,
            x=self.feature_cols[0],
            y=self.feature_cols[1],
            title="1. Empirical 2D Density Heatmap of Training Data"
        )
        fig.show()

    def plot_training_assignment(self):
        """2. Training Assignment Plot (Max Posterior Contour)"""
        xx, yy, grid = self._create_grid(self.X_train)

        # Calculate maximum posterior responsibility (\gamma_ik) across the grid
        probas = self.gmm.predict_proba(grid)
        max_probas = probas.max(axis=1).reshape(xx.shape)

        fig = go.Figure()

        # Continuous contour of the max responsibility
        fig.add_trace(go.Contour(
            x=xx[0], y=yy[:, 0], z=max_probas,
            opacity=0.6
        ))

        # Overlay training data points
        preds = self.gmm.predict(self.X_train)
        fig.add_trace(go.Scatter(
            x=self.X_train[:, 0], y=self.X_train[:, 1],
            mode='markers',
            marker=dict(color=preds, size=4),
            name='Train Data'
        ))

        fig.update_layout(
            title="2. Training Assignment Plot (Soft Boundaries)",
            xaxis_title=f"{self.feature_cols[0]} (Scaled)",
            yaxis_title=f"{self.feature_cols[1]} (Scaled)"
        )
        fig.show()

    def plot_test_assignment(self):
        """3. Test Assignment Plot (Out-of-sample Generalization)"""
        xx, yy, grid = self._create_grid(self.X_test)

        probas = self.gmm.predict_proba(grid)
        max_probas = probas.max(axis=1).reshape(xx.shape)

        fig = go.Figure()

        # Background contour mapping learned from train applied to test range
        fig.add_trace(go.Contour(
            x=xx[0], y=yy[:, 0], z=max_probas,
            opacity=0.6
        ))

        # Overlay test data points
        preds = self.gmm.predict(self.X_test)
        fig.add_trace(go.Scatter(
            x=self.X_test[:, 0], y=self.X_test[:, 1],
            mode='markers',
            marker=dict(color=preds, size=5, symbol='diamond'),
            name='Test Data'
        ))

        fig.update_layout(
            title="3. Test Assignment Plot (Out-of-Sample Overlay)",
            xaxis_title=f"{self.feature_cols[0]} (Scaled)",
            yaxis_title=f"{self.feature_cols[1]} (Scaled)"
        )
        fig.show()

r'''
Visual Evaluation and The Conditional Expectation:
The continuous background contour directly visualizes the mathematical behavior
of the expectation vector E[Z_i | X_i = x_grid]. Instead of showing rigid,
hard boundaries, the contour maps the magnitude of the maximum posterior
responsibility (max_k \gamma_{ik}). Near the center of a cluster's Gaussian density,
the contour approaches 1.0, visually confirming absolute certainty. As you move outward
into the spatial margins between two cluster centers, the maximum responsibility drops
heavily toward 0.5 (or 0.33 for K=3). Overlaying the out-of-sample test points over
this probability field demonstrates that GMMs learn continuous density functions that
smoothly dictate soft assignment vectors for entirely unseen points.
'''

# --- Execution Example ---
print("Loading data and running GMM...")
df = pd.read_csv('CC GENERAL.csv')
features = ['PURCHASES', 'CREDIT_LIMIT']

segmenter = GMMFinancialSegmenter(n_components=3)
segmenter.prepare_data(df, features)
segmenter.fit_and_evaluate()
segmenter.plot_empirical_density()
segmenter.plot_training_assignment()
segmenter.plot_test_assignment()
```
