# Q3) Transformer Failiures

In a certain electricity distribution district, the number of emergency transformer failures reported in a single day is modeled as a discrete random variable $X$. Assume that $X$ takes values in the set $\{0,1,2,3\}$, with probabilities

$$
P(X=0)=0.50,\qquad P(X=1)=0.30,\qquad P(X=2)=0.15,\qquad P(X=3)=0.05.
$$

1. Construct a probability space $(\Omega,\mathfrak F,P)$ suitable for this model.
2. Define $X$ as a random variable on this space.
3. Determine the probability mass function of $X$.
4. Compute the expectation $E[X]$.
5. Compute the variance $\mathrm{Var}(X)$.
6. Find $P(X\ge 1)$ and $P(X\ge 2)$.
7. Briefly explain the practical meaning of the expectation and variance in this setting.



### 1. Probability Space $(\Omega,\mathfrak{F},P)$

To model this situation, we construct the probability space as follows:
*   **Sample Space ($\Omega$):** The set of all possible outcomes (number of failures).
    $$\Omega=\{0,1,2,3\}$$
*   **Event Space ($\mathfrak{F}$):** The $\sigma$-algebra representing all possible events. Since $\Omega$ is finite, we use the power set of $\Omega$ (all possible subsets).
    $$\mathfrak{F}=2^\Omega$$
*   **Probability Measure ($P$):** A function $P:\mathfrak{F}\to[0,1]$ such that for any event $A\in\mathfrak{F}$:
    $$P(A)=\sum_{\omega\in A}p(\omega)$$
    where $p(0)=0.50$, $p(1)=0.30$, $p(2)=0.15$, and $p(3)=0.05$.

### 2. Random Variable $X$

A random variable is a measurable function from the sample space to the real numbers, $X:\Omega\to\mathbb{R}$. In this straightforward model, $X$ simply maps each outcome to its numerical value:
$$X(\omega)=\omega\quad\text{for all }\omega\in\Omega$$

### 3. Probability Mass Function (PMF) of $X$

The probability mass function $p_X(x)=P(X=x)$ describes the probability of each discrete value:
$$p_X(x)=\begin{cases}0.50&\text{if }x=0\\0.30&\text{if }x=1\\0.15&\text{if }x=2\\0.05&\text{if }x=3\\0&\text{otherwise}\end{cases}$$

### 4. Expectation $E[X]$

The expectation (or expected value) is the probability-weighted average of all possible values:
$$E[X]=\sum_{x}xP(X=x)$$
$$E[X]=(0\times 0.50)+(1\times 0.30)+(2\times 0.15)+(3\times 0.05)$$
$$E[X]=0+0.30+0.30+0.15=0.75$$

### 5. Variance $\mathrm{Var}(X)$

First, compute the expected value of $X^2$, denoted as $E[X^2]$:
$$E[X^2]=\sum_{x}x^2P(X=x)$$
$$E[X^2]=(0^2\times 0.50)+(1^2\times 0.30)+(2^2\times 0.15)+(3^2\times 0.05)$$
$$E[X^2]=0+0.30+(4\times 0.15)+(9\times 0.05)$$
$$E[X^2]=0.30+0.60+0.45=1.35$$

Now, use the computational formula for variance $\mathrm{Var}(X)=E[X^2]-(E[X])^2$:
$$\mathrm{Var}(X)=1.35-(0.75)^2$$
$$\mathrm{Var}(X)=1.35-0.5625=0.7875$$

### 6. Probabilities $P(X\ge 1)$ and $P(X\ge 2)$

**For $P(X\ge 1)$:**
Using the complement rule:
$$P(X\ge 1)=1-P(X=0)$$
$$P(X\ge 1)=1-0.50=0.50$$

**For $P(X\ge 2)$:**
Summing the probabilities of the specific outcomes:
$$P(X\ge 2)=P(X=2)+P(X=3)$$
$$P(X\ge 2)=0.15+0.05=0.20$$

### 7. Practical Meaning of Expectation and Variance

*   **Expectation:** Over a long period, the distribution district can anticipate an average of 0.75 emergency transformer failures per day. This metric is useful for long-term resource allocation, inventory management, and budget planning.
*   **Variance:** This measures the daily spread or volatility of transformer failures around the average. A variance of 0.7875 indicates day-to-day unpredictability, meaning maintenance crews should be prepared for days with multiple failures despite the relatively low daily average.

# Q4) Probability of Defective Items

In a manufacturing process, the proportion of defective items in a large batch produced on a given day is modeled by a continuous random variable $X$. Since $X$ represents a proportion, it takes values in the interval $[0,1]$. Suppose that $X$ has a Beta distribution with parameters $\alpha=2$ and $\beta=18$, so that its density is

$$
f(x)=
\begin{cases}
\dfrac{1}{B(2,18)}x^{1}(1-x)^{17}, & 0<x<1,\\\
0, & \text{otherwise},
\end{cases}
$$

where $B(2,18)$ is the Beta function.

Answer the following:

1. Construct a probability space $(\Omega,\mathfrak F,P)$ suitable for this model.

2. Define $X$ as a continuous random variable on this space.

3. Write down the probability density function of $X$.

4. Verify that $f$ is a valid probability density function.

5. Compute the expectation $E[X]$.

6. Compute the variance $\mathrm{Var}(X)$.

7. Find the probability that the defective proportion is at most $0.10$, that is, compute

$$
P(X\le 0.10).
$$

You may leave your answer in Beta integral form.

8. Find the probability that the defective proportion lies between $0.05$ and $0.15$, that is, compute

$$
P(0.05\le X\le 0.15).
$$

You may leave your answer in integral form.

9. Briefly explain why a Beta distribution is a reasonable model for this situation.

10. Interpret the expectation and variance in the context of production quality.


### 1. Probability Space $(\Omega,\mathfrak{F},P)$

To model a continuous proportion, we construct the probability space as follows:
*   **Sample Space ($\Omega$):** The set of all possible proportions of defective items, which is the closed interval from 0 to 1.
    $$\Omega=[0,1]$$
*   **Event Space ($\mathfrak{F}$):** The Borel $\sigma$-algebra on $[0,1]$, denoted as $\mathcal{B}([0,1])$. This collection includes all intervals and their countable unions/intersections within $[0,1]$.
*   **Probability Measure ($P$):** Defined for any event $A\in\mathfrak{F}$ by integrating the density function over that event:
    $$P(A)=\int_{A} f(x)\,dx$$

### 2. Random Variable $X$

The continuous random variable $X$ is defined as the identity function mapping a specific observed proportion to its numerical value:
$$X:\Omega\to\mathbb{R}$$
$$X(\omega)=\omega\quad\text{for all }\omega\in\Omega$$

### 3. Probability Density Function (PDF) of $X$

The probability density function is given by the Beta distribution with parameters $\alpha=2$ and $\beta=18$:
$$f(x)=\begin{cases}\dfrac{1}{B(2,18)}x(1-x)^{17}&\text{if }0<x<1\\0&\text{otherwise}\end{cases}$$

### 4. Verification of Valid PDF

To be a valid PDF, $f(x)$ must satisfy two conditions:
1.  **Non-negativity:** For $0<x<1$, $x>0$ and $(1-x)^{17}>0$. Since $B(2,18)>0$, $f(x)\ge 0$ everywhere.
2.  **Integrates to 1:** We evaluate the integral over the entire domain:
    $$\int_{-\infty}^{\infty}f(x)\,dx=\int_{0}^{1}\frac{1}{B(2,18)}x^{2-1}(1-x)^{18-1}\,dx$$
    By the definition of the Beta function, $\int_{0}^{1}x^{\alpha-1}(1-x)^{\beta-1}\,dx = B(\alpha,\beta)$. Thus:
    $$\int_{-\infty}^{\infty}f(x)\,dx=\frac{1}{B(2,18)}\cdot B(2,18)=1$$
    Both conditions are met.

### 5. Expectation $E[X]$

For a Beta distribution with parameters $\alpha$ and $\beta$, the expected value is given by the formula $E[X]=\frac{\alpha}{\alpha+\beta}$.
$$E[X]=\frac{2}{2+18}=\frac{2}{20}=0.10$$

### 6. Variance $\mathrm{Var}(X)$

For a Beta distribution, the variance is given by $\mathrm{Var}(X)=\frac{\alpha\beta}{(\alpha+\beta)^2(\alpha+\beta+1)}$.
$$\mathrm{Var}(X)=\frac{2\times 18}{(2+18)^2(2+18+1)}$$
$$\mathrm{Var}(X)=\frac{36}{(20)^2(21)}=\frac{36}{400\times 21}=\frac{36}{8400}=\frac{3}{700}\approx 0.004286$$

### 7. Probability $P(X\le 0.10)$

To find the probability that the proportion is at most 0.10, we integrate the PDF from 0 to 0.10. Leaving it in integral form:
$$P(X\le 0.10)=\frac{1}{B(2,18)}\int_{0}^{0.10}x(1-x)^{17}\,dx$$
*(Note: This is also known as the regularized incomplete beta function, $I_{0.10}(2,18)$).*

### 8. Probability $P(0.05\le X\le 0.15)$

To find the probability that the defective proportion lies strictly between 0.05 and 0.15, we set those as our integral bounds:
$$P(0.05\le X\le 0.15)=\frac{1}{B(2,18)}\int_{0.05}^{0.15}x(1-x)^{17}\,dx$$

### 9. Why a Beta Distribution is Reasonable

The Beta distribution is the standard mathematical model for variables that represent proportions or probabilities. Because its domain is strictly constrained to the interval $[0,1]$, it perfectly mirrors the real-world constraints of a percentage (you cannot have less than 0% or more than 100% defective items). Furthermore, by adjusting $\alpha$ and $\beta$, the curve can be heavily skewed (as it is here) to represent highly specialized continuous processes.

### 10. Interpretation of Expectation and Variance

*   **Expectation (0.10):** Over the long run, the factory can expect that on an average day, exactly 10% of the manufactured batch will be defective.
*   **Variance ($\approx 0.0043$):** This very small variance indicates that the daily proportion of defects is highly concentrated around that 10% average. The manufacturing process is consistent, meaning it is highly unlikely to have a day with, for example, 40% defects or 0% defects.

# Q5) Expected Project Completion Time

A construction company is estimating the time required to install a backup generator at a hospital. Let $X$ denote the number of hours needed to complete the installation. Based on expert judgment, the shortest possible time is $4$ hours, the longest possible time is $10$ hours, and the most likely completion time is $6$ hours. Assume that $X$ follows a Triangular distribution with parameters $a=4$, $b=10$, and $c=6$.

Answer the following:

1. Specify the support of $X$.

2. Write down the probability density function $f_X(x)$.

3. Verify that $f_X$ integrates to $1$ over its support.

4. Compute the expectation $E[X]$.

5. Compute the variance $\mathrm{Var}(X)$.

6. Interpret the values of $a$, $b$, and $c$ in the context of this problem.

7. Explain briefly why a Triangular distribution is a reasonable model in this setting.

8. Compute the probability that the installation takes at most $5$ hours, that is, compute

$$
P(X \leq 5).
$$

9. Compute the probability that the installation takes between $5$ and $8$ hours, that is, compute

$$
P(5 \leq X \leq 8).
$$

10. Briefly explain how this model could be used in project planning or sensitivity analysis.


### 1. Support of $X$

The support of a Triangular distribution is the interval of all possible values it can take. Since the shortest time is 4 and the longest is 10, the support is the closed interval:
$$[4, 10]$$

### 2. Probability Density Function $f_X(x)$

For a Triangular distribution with minimum $a=4$, maximum $b=10$, and mode $c=6$, the probability density function is defined piece-wise:

$$
f_X(x) = 
\begin{cases} 
\dfrac{2(x-a)}{(b-a)(c-a)}, & a \le x < c \\[10pt]
\dfrac{2(b-x)}{(b-a)(b-c)}, & c \le x \le b \\[10pt]
0, & \text{otherwise} 
\end{cases}
$$

Substituting the parameters $a=4$, $b=10$, $c=6$:
*   For $4 \le x < 6$: $\dfrac{2(x-4)}{(10-4)(6-4)} = \dfrac{2(x-4)}{6 \times 2} = \dfrac{x-4}{6}$
*   For $6 \le x \le 10$: $\dfrac{2(10-x)}{(10-4)(10-6)} = \dfrac{2(10-x)}{6 \times 4} = \dfrac{10-x}{12}$

So the PDF is:
$$
f_X(x) = 
\begin{cases} 
\dfrac{x-4}{6}, & 4 \le x < 6 \\[10pt]
\dfrac{10-x}{12}, & 6 \le x \le 10 \\[10pt]
0, & \text{otherwise} 
\end{cases}
$$

### 3. Verify Integration to 1

We can verify this geometrically or using calculus. Geometrically, the PDF forms a triangle with base $b-a$ and peak height $\frac{2}{b-a}$ at $x=c$. 
*   Base $= 10 - 4 = 6$
*   Height $= \frac{2}{10-4} = \frac{2}{6} = \frac{1}{3}$
*   Area $= \frac{1}{2} \times \text{base} \times \text{height} = \frac{1}{2} \times 6 \times \frac{1}{3} = 1$.

Using calculus, we integrate the PDF over the support:
$$ \int_{4}^{10} f_X(x)\,dx = \int_{4}^{6} \frac{x-4}{6}\,dx + \int_{6}^{10} \frac{10-x}{12}\,dx $$
$$ = \left[ \frac{(x-4)^2}{12} \right]_{4}^{6} + \left[ -\frac{(10-x)^2}{24} \right]_{6}^{10} $$
$$ = \left( \frac{4}{12} - 0 \right) + \left( 0 - \left(-\frac{16}{24}\right) \right) = \frac{1}{3} + \frac{2}{3} = 1 $$

### 4. Expectation $E[X]$

The expected value formula for a Triangular distribution is the simple average of the three parameters:
$$E[X] = \frac{a+b+c}{3}$$
$$E[X] = \frac{4+10+6}{3} = \frac{20}{3} \approx 6.67 \text{ hours}$$

### 5. Variance $\mathrm{Var}(X)$

The formula for the variance of a Triangular distribution is:
$$\mathrm{Var}(X) = \frac{a^2 + b^2 + c^2 - ab - ac - bc}{18}$$
$$\mathrm{Var}(X) = \frac{4^2 + 10^2 + 6^2 - (4\times10) - (4\times6) - (10\times6)}{18}$$
$$\mathrm{Var}(X) = \frac{16 + 100 + 36 - 40 - 24 - 60}{18}$$
$$\mathrm{Var}(X) = \frac{152 - 124}{18} = \frac{28}{18} = \frac{14}{9} \approx 1.556$$

### 6. Interpretation of $a$, $b$, and $c$

*   **$a=4$ (Optimistic Time):** The absolute fastest the installation can be completed if everything goes perfectly.
*   **$b=10$ (Pessimistic Time):** The absolute longest the installation will take if there are significant delays or complications.
*   **$c=6$ (Most Likely Time):** The standard, expected time to finish under normal conditions. 

### 7. Why a Triangular Distribution is Reasonable

In real-world project planning, it is often impossible to have enough historical data to fit a complex curve like a Normal or Gamma distribution. The Triangular distribution relies entirely on expert intuition. An experienced contractor can easily estimate "best case," "worst case," and "normal case" scenarios, making this model practical and highly effective for modeling human-estimated task durations.

### 8. Probability $P(X \le 5)$

Since $5$ is less than the mode ($c=6$), we integrate the first part of the PDF from 4 to 5 (or use the CDF formula $\frac{(x-a)^2}{(b-a)(c-a)}$):
$$P(X \le 5) = \int_{4}^{5} \frac{x-4}{6}\,dx = \left[ \frac{(x-4)^2}{12} \right]_{4}^{5}$$
$$P(X \le 5) = \frac{(5-4)^2}{12} - 0 = \frac{1}{12} \approx 0.0833$$

### 9. Probability $P(5 \le X \le 8)$

We can find this by subtracting cumulative probabilities: $P(X \le 8) - P(X \le 5)$. 
Since $8 > c$, we use the upper CDF formula to find $P(X \le 8)$:
$$P(X \le 8) = 1 - \frac{(b-x)^2}{(b-a)(b-c)} = 1 - \frac{(10-8)^2}{(10-4)(10-6)} = 1 - \frac{4}{24} = 1 - \frac{1}{6} = \frac{5}{6}$$
Now, subtract $P(X \le 5)$ which we found in Part 8:
$$P(5 \le X \le 8) = \frac{5}{6} - \frac{1}{12} = \frac{10}{12} - \frac{1}{12} = \frac{9}{12} = \frac{3}{4} = 0.75$$

### 10. Use in Project Planning / Sensitivity Analysis

This model is a core component of PERT (Project Evaluation and Review Technique) planning. It translates subjective expert guesses into objective probabilities. Instead of just giving a client a single number ("It'll take 6.67 hours"), a manager can assess risk and promise deadlines with confidence bounds, such as knowing there is exactly a 75% chance the job will take between 5 and 8 hours.

# Q6) Expected Project Completion Time Using PERT

A project manager is estimating the time required to complete a critical software integration task. Let $X$ denote the number of days required to complete the task. Based on expert judgment, the shortest possible time is $a=8$ days, the longest possible time is $b=20$ days, and the most likely completion time is $c=11$ days. Assume that $X$ follows a PERT distribution with parameters $(a,b,c)$.

The PERT distribution is defined as a Beta distribution scaled to the interval $[a,b]$, with internal shape parameters

$$
\alpha = 1 + 4\left( \frac{c-a}{b-a} \right)
$$

and

$$
\beta = 1 + 4\left( \frac{b-c}{b-a} \right).
$$

Its probability density function is

$$
f(x)=\frac{(x-a)^{\alpha-1}(b-x)^{\beta-1}}{B(\alpha,\beta)(b-a)^{\alpha+\beta-1}}, \qquad a \leq x \leq b,
$$

where

$$
B(\alpha,\beta)=\int_0^1 t^{\alpha-1}(1-t)^{\beta-1},dt.
$$

Answer the following:

1. Compute the values of the shape parameters $\alpha$ and $\beta$.

2. Specify the support of $X$.

3. Write down the probability density function $f_X(x)$ for this model.

4. State the probability space $(\Omega,\mathfrak F,P)$ suitable for this continuous model.

5. Define $X$ as a random variable on this space.

6. Show that the expectation $E[X]$ is given by

$$
E[X]=\frac{a+4c+b}{6}.
$$

7. Compute the variance $\mathrm{Var}(X)$ using the approximation

$$
\mathrm{Var}(X)=\frac{(E[X]-a)(b-E[X])}{7}.
$$

8. Compute the approximate standard deviation $\sigma$.

9. Briefly explain why a PERT distribution may be more appropriate than a Triangular distribution in this setting byt comparing the two results.

10. Interpret the practical meaning of the values $a$, $b$, $c$, $E[X]$, and $\sigma$ in the context of project planning.


### 1. Shape Parameters $\alpha$ and $\beta$

Given $a=8$, $b=20$, and $c=11$, we first calculate the span $(b-a)=12$.
$$ \alpha = 1 + 4\left( \frac{11-8}{20-8} \right) = 1 + 4\left( \frac{3}{12} \right) = 1 + 4(0.25) = 2 $$
$$ \beta = 1 + 4\left( \frac{20-11}{20-8} \right) = 1 + 4\left( \frac{9}{12} \right) = 1 + 4(0.75) = 4 $$

### 2. Support of $X$

The support of the continuous random variable $X$ is the bounded interval between the optimistic and pessimistic times:
$$[8, 20]$$

### 3. Probability Density Function $f_X(x)$

Substitute the parameters $a=8$, $b=20$, $\alpha=2$, and $\beta=4$ into the given PDF. Note that $(b-a)^{\alpha+\beta-1} = 12^{2+4-1} = 12^5$.
$$
f_X(x) = 
\begin{cases} 
\dfrac{(x-8)^{1}(20-x)^{3}}{B(2,4) \cdot 12^5}, & 8 \le x \le 20 \\[10pt]
0, & \text{otherwise}
\end{cases}
$$

### 4. Probability Space $(\Omega,\mathfrak{F},P)$

*   **Sample Space ($\Omega$):** The interval of possible completion days, $\Omega = [8, 20]$.
*   **Event Space ($\mathfrak{F}$):** The Borel $\sigma$-algebra on $[8, 20]$, denoted $\mathcal{B}([8, 20])$.
*   **Probability Measure ($P$):** Defined for any event $A \in \mathfrak{F}$ as $P(A) = \int_A f_X(x)\,dx$.

### 5. Random Variable $X$

$X$ is defined as the identity mapping from the sample space to the real numbers:
$$X: \Omega \to \mathbb{R}$$
$$X(\omega) = \omega \quad \text{for all } \omega \in [8, 20]$$

### 6. Show Expectation $E[X]$

A PERT distribution is a shifted and scaled Beta distribution. If $Y \sim \text{Beta}(\alpha, \beta)$, then $E[Y] = \frac{\alpha}{\alpha+\beta}$.
The variable $X$ maps this to the interval $[a,b]$: $X = a + (b-a)Y$.
First, note that $\alpha+\beta = \left(1+4\frac{c-a}{b-a}\right) + \left(1+4\frac{b-c}{b-a}\right) = 2 + 4\left(\frac{b-a}{b-a}\right) = 6$.
Now, take the expectation of $X$:
$$ E[X] = a + (b-a)E[Y] = a + (b-a)\left(\frac{\alpha}{6}\right) $$
Substitute $\alpha$:
$$ E[X] = a + \frac{(b-a)\left(1 + 4\frac{c-a}{b-a}\right)}{6} = a + \frac{(b-a) + 4(c-a)}{6} $$
$$ E[X] = \frac{6a + b - a + 4c - 4a}{6} = \frac{a + 4c + b}{6} $$
*(For this specific problem, $E[X] = \frac{8 + 44 + 20}{6} = \frac{72}{6} = 12$.)*

### 7. Variance $\mathrm{Var}(X)$

Using the provided approximation formula and $E[X]=12$:
$$ \mathrm{Var}(X) = \frac{(E[X]-a)(b-E[X])}{7} $$
$$ \mathrm{Var}(X) = \frac{(12-8)(20-12)}{7} = \frac{4 \times 8}{7} = \frac{32}{7} \approx 4.571 $$

### 8. Standard Deviation $\sigma$

The standard deviation is the square root of the variance:
$$ \sigma = \sqrt{\frac{32}{7}} \approx 2.138 \text{ days} $$

### 9. PERT vs. Triangular Distribution

While both distributions use the same three parameters ($a, b, c$), the PERT distribution creates a smooth, bell-shaped curve rather than rigid straight lines. This concentrates more probability mass around the "most likely" value (mode) and artificially reduces the probability of extreme values at the tails. In real-world software development, tasks usually cluster closely around the expected time, making PERT's "bell curve" assumption more realistic than the Triangular model's linear drop-off.

### 10. Interpretation in Project Planning

*   **$a=8$:** The optimistic time; the project won't finish faster than 8 days.
*   **$b=20$:** The pessimistic time; assuming everything goes wrong, it caps at 20 days.
*   **$c=11$:** The mode; the single most common, "normal" scenario.
*   **$E[X]=12$:** The expected average. Notice it is higher than the mode ($11$). Because the worst-case scenario (20) stretches further than the best-case scenario (8), the "average" risk is pulled slightly to the right. 
*   **$\sigma \approx 2.14$:** The uncertainty spread. Planners can confidently expect the task to finish within roughly $12 \pm 2.14$ days.

# Q7) If the PERT Expectation is to Match Our Intuition

A project manager models the duration $X$ of a task by a PERT distribution with parameters $a$, $b$, and $c$, where $a$ is the minimum time, $b$ is the maximum time, and $c$ is the most likely time.

Suppose the minimum time is $a=5$ days and the most likely time is $c=11$ days. If the expected time computed from the PERT model is also to be the same as the most likely time:

1. Determine the value of the maximum time $b$.

2. What happens to the variance?

### 1. Determine the Maximum Time $b$

We are given that the expected time $E[X]$ is equal to the most likely time $c$. 
In a PERT distribution, the expectation is defined as:
$$E[X] = \frac{a + 4c + b}{6}$$

Since $E[X] = c$, we can set up the equation:
$$c = \frac{a + 4c + b}{6}$$

Multiply both sides by 6:
$$6c = a + 4c + b$$

Subtract $4c$ from both sides:
$$2c = a + b$$

Now, substitute the given values $a = 5$ and $c = 11$:
$$2(11) = 5 + b$$
$$22 = 5 + b$$
$$b = 17$$

The maximum time $b$ is **$17$ days**.

### 2. What Happens to the Variance?

Because $2c = a + b$, or equivalently $c = \frac{a+b}{2}$, the mode ($c=11$) sits exactly halfway between the minimum ($a=5$) and the maximum ($b=17$). 

**1. The Distribution Becomes Perfectly Symmetrical**
Because the mode is perfectly centered, the PERT distribution loses all skewness. The internal shape parameters become identical:
$$ \alpha = 1 + 4\left(\frac{11-5}{17-5}\right) = 1 + 4\left(\frac{6}{12}\right) = 3 $$
$$ \beta = 1 + 4\left(\frac{17-11}{17-5}\right) = 1 + 4\left(\frac{6}{12}\right) = 3 $$
Since $\alpha = \beta = 3$, the resulting curve looks perfectly balanced, heavily resembling a standard Normal distribution (bell curve).

**2. Calculating the Variance**
We can compute the variance in two ways:

*   **Exact Variance (Beta distribution formula):**
    For a Beta distribution scaled to $[a, b]$, the exact variance when $\alpha=\beta=3$ simplifies to:
    $$\mathrm{Var}(X) = \frac{(b-a)^2}{28}$$
    $$\mathrm{Var}(X) = \frac{(17-5)^2}{28} = \frac{144}{28} = \frac{36}{7} \approx 5.143$$

*   **Standard PERT Heuristic Variance:**
    Often in project management, the PERT variance is approximated using $\left(\frac{b-a}{6}\right)^2$:
    $$\mathrm{Var}(X) = \frac{(17-5)^2}{36} = \frac{144}{36} = 4$$

In short, the variance reflects a perfectly symmetric, unskewed spread of uncertainty around the 11-day average.
