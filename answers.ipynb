# Q1) The Probability of Shooting at a Target

A point is chosen at random from a disk of radius $10$. Let $A$ be the event that the point lies within $1$ unit of the boundary.

1. Model the experiment assuming the point is uniformly distributed with respect to area, and compute $P(A)$.
2. Model the experiment assuming the distance to the point from the center is chosen uniformly from $[0,10]$, and the direction is chosen independently and uniformly from $[0,2\pi)$ and compute $P(A)$.
3. Explain why the two answers differ.


# Probability in a Disk

A point is chosen at random from a disk of radius $10$. Let $A$ be the event that the point lies within $1$ unit of the boundary.

---

## 1.Uniform with respect to area

A uniform distribution over the disk means probability is proportional to area.

Event:
$$
A = \{9 \le r \le 10\}
$$

Total area:
$$
A_{\text{total}} = \pi (10)^2 = 100\pi
$$

Inner disk area:
$$
A_{\text{inner}} = \pi (9)^2 = 81\pi
$$

Annulus area:
$$
A_A = 100\pi - 81\pi = 19\pi
$$

Probability:
$$
P(A) = \frac{19\pi}{100\pi} = \frac{19}{100} = 0.19
$$



## 2.Uniform radius and direction

Here:
- $r \sim \text{Uniform}[0,10]$
- $\theta \sim \text{Uniform}[0,2\pi)$

Event:
$$
A = \{9 \le r \le 10\}
$$

Probability:
$$
P(A) = \frac{10 - 9}{10} = \frac{1}{10} = 0.1
$$



## 3.Explanation

The two answers differ because the distributions are different.

In polar coordinates:
$$
dA = r \, dr \, d\theta
$$

So for a uniform area distribution, the density depends on $r$.  
A uniform radius distribution ignores this, giving too much weight to the center.



## Final Answers

$$
\boxed{P(A) = 0.19 \text{ (uniform area)}}
$$

$$
\boxed{P(A) = 0.10 \text{ (uniform radius)}}
$$

# Q2) Bertrand's Paradox

A chord of a circle is chosen “at random.” What is the probability that its length exceeds the side length of the inscribed equilateral triangle?

Show that different mathematically natural models of “random chord” lead to different answers.

# Bertrands Paradox: Random Chord Problem

A chord of a circle is chosen “at random.” What is the probability that its length exceeds the side length of the inscribed equilateral triangle?



## Fact

For a circle of radius $R$, the side length of the inscribed equilateral triangle is:
$$
s = \sqrt{3}\,R
$$

A chord at distance $d$ from the center has length:
$$
L = 2\sqrt{R^2 - d^2}
$$

We want:
$$
L > \sqrt{3}R
$$



## Solve the inequality

$$
2\sqrt{R^2 - d^2} > \sqrt{3}R
$$

Square both sides:
$$
4(R^2 - d^2) > 3R^2
$$

$$
4R^2 - 4d^2 > 3R^2
$$

$$
R^2 > 4d^2
$$

$$
d < \frac{R}{2}
$$

So the event occurs when the chord is within distance $R/2$ from the center.



## Model1:Random midpoint uniformly in the disk

The chord is determined by choosing its midpoint uniformly in the circle.

Probability is proportional to area:
$$
P = \frac{\pi (R/2)^2}{\pi R^2} = \frac{1}{4}
$$



## Model2:Random radius, perpendicular chord

Choose a random radius and a random point uniformly along it, then draw a perpendicular chord.

Here $d \sim \text{Uniform}[0,R]$

$$
P(d < R/2) = \frac{R/2}{R} = \frac{1}{2}
$$



## Model3:Random endpoints on the circumference

Pick two random points independently on the circle.

This leads to:
$$
P = \frac{1}{3}
$$



## Final Answers

$$
\boxed{P = \frac{1}{4}} \quad \text{(random midpoint)}
$$

$$
\boxed{P = \frac{1}{2}} \quad \text{(random radius)}
$$

$$
\boxed{P = \frac{1}{3}} \quad \text{(random endpoints)}
$$



## Conclusion

Different mathematically natural definitions of “random chord” produce different probabilities.

This is known as **Bertrand’s Paradox**: the phrase “at random” is ambiguous unless the method of randomization is clearly specified.

# Q8) Probability of Being a Girl?

A family has two children. Assume each child is equally likely to be a boy or a girl, independently of the other.

What is the probability that both children are girls, given that **at least one** of them is a girl?

# TwoChildrenProblem

A family has two children. Each child is equally likely to be a boy (B) or a girl (G), independently.

We want:
$$
P(\text{both are girls} \mid \text{at least one is a girl})
$$

---

## Sample Space

All equally likely outcomes:
$$
\{BB, BG, GB, GG\}
$$

---

## Condition on "at least one girl"

Exclude $BB$:
$$
\{BG, GB, GG\}
$$

---

## Favorable outcomes

Only $GG$ satisfies "both are girls".

---

## Compute probability

$$
P = \frac{1}{3}
$$

---

## Final Answer

$$
\boxed{\frac{1}{3}}
$$

### Solution

Assume each child is equally likely to be a boy (B) or a girl (G), independently.

The sample space for two children is:
\[
\{BB, BG, GB, GG\}
\]
Each outcome has probability:
\[
\frac{1}{4}.
\]

---

### Step 1: Condition on “at least one is a girl”

The event “at least one girl” excludes the outcome $BB$.

So the conditional sample space becomes:
\[
\{BG, GB, GG\}
\]

Each of these outcomes is now equally likely with probability:
\[
\frac{1}{3}.
\]

---

### Step 2: Find the required probability

We want:
\[
P(\text{both girls} \mid \text{at least one girl})
\]

Only one outcome satisfies “both girls,” which is $GG$.

Thus,
\[
P = \frac{1}{3}.
\]

---

### Final Answer

\[
\boxed{\frac{1}{3}}
\]

# Q9) Information Available in a Discrete Random Variable

A factory inspects a product and records its quality using two different coding systems.

Each product falls into exactly one of the following four categories:

* $G_1$: good quality, produced on Machine 1
* $G_2$: good quality, produced on Machine 2
* $D_1$: defective, produced on Machine 1
* $D_2$: defective, produced on Machine 2

Assume the sample space is

$$
\Omega=\{G_1,G_2,D_1,D_2\},
$$

with probabilities

$$
P(\{G_1\})=0.50,\qquad P(\{G_2\})=0.20,\qquad P(\{D_1\})=0.10,\qquad P(\{D_2\})=0.20.
$$

Define two discrete random variables on the same probability space:

* $X$ records only whether the item is good or defective:
  $$
  X(G_1)=0,\quad X(G_2)=0,\quad X(D_1)=1,\quad X(D_2)=1.
  $$

* $Y$ records both quality and machine:
  $$
  Y(G_1)=1,\quad Y(G_2)=2,\quad Y(D_1)=3,\quad Y(D_2)=4.
  $$

Answer the following.

1. Write down the probability space $(\Omega,\mathfrak F,P)$, where $\mathfrak F=\mathfrak P(\Omega)$.

2. Compute the $\sigma$-algebra generated by $X$, namely $\sigma(X)$.

3. Compute the $\sigma$-algebra generated by $Y$, namely $\sigma(Y)$.

4. Show that $\sigma(X)\subseteq \sigma(Y)$, and explain why $Y$ contains more information than $X$.

5. Find the marginal distribution of $X$.

6. Find the marginal distribution of $Y$.

7. Compute the joint probabilities
   $$
   P(X=0,Y=1),\quad P(X=0,Y=2),\quad P(X=1,Y=3),\quad P(X=1,Y=4).
   $$

8. Compute the conditional probabilities
   $$
   P(X=1\mid Y=3),\qquad P(X=1\mid Y=4),\qquad P(Y=3\mid X=1),\qquad P(Y=4\mid X=1).
   $$

9. Interpret the difference between conditioning on $X$ and conditioning on $Y$.

10. Explain, in words, what information is available in $\sigma(X)$ but not in $\sigma(Y)$, and what information is available in $\sigma(Y)$ but not in $\sigma(X)$.

11. Give a Python based visualization of the model.

12. Find the marginal, joint, and conditional Shannon Entropy of $X$ and $Y$.

# Probability Space and Random Variables

## 1.Probability Space

The sample space is:

$$
\Omega = \{G_1, G_2, D_1, D_2\}
$$

The sigma-algebra is:

$$
\mathfrak{F} = \mathcal{P}(\Omega)
$$

The probability measure is:

$$
P(G_1)=0.50,\quad P(G_2)=0.20,\quad P(D_1)=0.10,\quad P(D_2)=0.20
$$



## 2.Random Variables

### Random variable \(X\)

$$
X(G_1)=0,\quad X(G_2)=0,\quad X(D_1)=1,\quad X(D_2)=1
$$

### Random variable \(Y\)

$$
Y(G_1)=1,\quad Y(G_2)=2,\quad Y(D_1)=3,\quad Y(D_2)=4
$$



## 3.Sigma-algebra generated by \(X\)

Since \(X\) groups outcomes into good and defective:

$$
\sigma(X)=\{\emptyset,\Omega,\{G_1,G_2\},\{D_1,D_2\}\}
$$



## 4.Sigma-algebra generated by \(Y\)

Since \(Y\) uniquely identifies each outcome:

$$
\sigma(Y)=\mathcal{P}(\Omega)
$$



## 5Relationship between \(\sigma(X)\) and \(\sigma(Y)\)

$$
\sigma(X)\subseteq \sigma(Y)
$$

Thus, \(Y\) contains more detailed information than \(X\).



## 6.Marginal Distribution of \(X\)

$$
P(X=0)=P(G_1)+P(G_2)=0.7
$$

$$
P(X=1)=P(D_1)+P(D_2)=0.3
$$



## 7.Marginal Distribution of \(Y\)

$$
P(Y=1)=0.50,\quad P(Y=2)=0.20,\quad P(Y=3)=0.10,\quad P(Y=4)=0.20
$$



## 8.Joint Probabilities

$$
P(X=0,Y=1)=0.50
$$

$$
P(X=0,Y=2)=0.20
$$

$$
P(X=1,Y=3)=0.10
$$

$$
P(X=1,Y=4)=0.20
$$

---

## 9.Conditional Probabilities

$$
P(X=1|Y=3)=1,\quad P(X=1|Y=4)=1
$$

$$
P(Y=3|X=1)=\frac{0.10}{0.30}=\frac{1}{3}
$$

$$
P(Y=4|X=1)=\frac{0.20}{0.30}=\frac{2}{3}
$$



## 10. Interpretation

- \(X\): only tells whether product is good or defective.
- \(Y\): gives full information (quality + machine).

Thus, \(Y\) is more informative than \(X\).



## 11. Entropy Calculations

### Entropy of \(X\)

$$
H(X) = -0.7\log_2(0.7) - 0.3\log_2(0.3)
$$

$$
H(X) \approx 0.881 \text{ bits}
$$



### Entropy of \(Y\)

$$
H(Y) = -\sum p(y)\log_2 p(y)
$$

$$
H(Y)=-(0.5\log_2 0.5 + 0.2\log_2 0.2 + 0.1\log_2 0.1 + 0.2\log_2 0.2)
$$

$$
H(Y) \approx 1.761 \text{ bits}
$$



### Joint Entropy

Since mapping is one-to-one:

$$
H(X,Y)=H(Y)\approx 1.761 \text{ bits}
$$



### Conditional Entropy

$$
H(X|Y)=0
$$

$$
H(Y|X)\approx 0.881 \text{ bits}
$$



## 12. Summary

$$
H(X)=0.881,\quad H(Y)=1.761
$$

$$
H(X|Y)=0,\quad H(Y|X)=0.881
$$
